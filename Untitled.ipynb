```python
#!/usr/bin/env python3
# MODIS NDVI (MOD13Q1.061) via AppEEARS for Iowa, last 2 years
# Handles 200/201/202, normalizes relative Location headers, and converts
# /api/task/{id} → /api/bundle/{id} before listing files.

import os
import time
import json
import zipfile
import getpass
from datetime import date, timedelta
from urllib.parse import urljoin

import requests
from tqdm import tqdm

API_ROOT = "https://appeears.earthdatacloud.nasa.gov"
API = f"{API_ROOT}/api"
OUT_DIR = "appeears_ndvi_iowa"
os.makedirs(OUT_DIR, exist_ok=True)

# Iowa bbox (W,S,E,N)
BBOX = (-96.6395, 40.3754, -90.1401, 43.5012)

END = date.today()
START = END - timedelta(days=365*2)

def mmddyyyy(d: date) -> str:
    return d.strftime("%m-%d-%Y")

def login(username: str, password: str) -> str:
    r = requests.post(f"{API}/login", auth=(username, password))
    r.raise_for_status()
    tok = r.json().get("token")
    if not tok:
        raise SystemExit("No token returned from /login.")
    return tok

def get_ndvi_layer_name():
    product_id = "MOD13Q1.061"
    r = requests.get(f"{API}/product/{product_id}")
    r.raise_for_status()
    data = r.json()
    layers = []
    if isinstance(data, dict):
        for _, v in data.items():
            if isinstance(v, dict) and "Layer" in v:
                layers.append(v)
    elif isinstance(data, list):
        layers = data
    names = [L["Layer"] for L in layers if "Layer" in L]
    prefer = [n for n in names if "ndvi" in n.lower() and "250m" in n.lower()]
    return product_id, (prefer[0] if prefer else next(n for n in names if "ndvi" in n.lower()))

def featurecollection_from_bbox(bbox):
    w, s, e, n = bbox
    return {
        "type": "FeatureCollection",
        "features": [{
            "type": "Feature",
            "properties": {"name": "Iowa_bbox"},
            "geometry": {
                "type": "Polygon",
                "coordinates": [[[w, s], [e, s], [e, n], [w, n], [w, s]]]
            }
        }]
    }

def normalize_bundle_url(path_or_url: str, task_id_hint: str | None = None) -> str:
    """
    - If relative, prepend API_ROOT.
    - If it points at /task/{id}, convert to /bundle/{id}.
    """
    url = path_or_url.strip()
    if not url.lower().startswith("http"):
        url = urljoin(API_ROOT, url)  # make absolute
    # Convert .../task/{id} → .../bundle/{id}
    # (Some deployments return a task URL after completion; we need bundle.)
    if "/task/" in url and "/bundle/" not in url:
        tid = url.rstrip("/").rsplit("/", 1)[-1]
        if not tid and task_id_hint:
            tid = task_id_hint
        url = f"{API}/bundle/{tid}"
    return url

def submit_area_task(token: str, task_name: str, product_id: str, layer_name: str,
                     start: date, end: date, bbox) -> str:
    headers = {"Authorization": f"Bearer {token}", "Content-Type": "application/json"}
    payload = {
        "task_type": "area",
        "task_name": task_name,
        "params": {
            "dates": [{"startDate": mmddyyyy(start), "endDate": mmddyyyy(end)}],
            "layers": [{"product": product_id, "layer": layer_name}],
            "geo": featurecollection_from_bbox(bbox),
            "output": {
                "format": {"type": "geotiff"},
                "projection": "geographic"
            }
        }
    }
    r = requests.post(f"{API}/task", headers=headers, data=json.dumps(payload))
    if r.status_code not in (200, 201, 202):
        raise SystemExit(f"Task submission failed ({r.status_code}): {r.text}")
    task_id = None
    try:
        j = r.json()
        task_id = j.get("task_id") or j.get("taskID")
    except Exception:
        pass
    if not task_id:
        loc = r.headers.get("Location", "")
        if "/task/" in loc:
            task_id = loc.rstrip("/").rsplit("/", 1)[-1]
    if not task_id:
        raise SystemExit(f"Could not parse task_id from response ({r.status_code}). Body:\n{r.text}")
    return task_id

def poll_until_done(token: str, task_id: str, sleep_s: int = 15, max_wait_min: int = 120) -> str:
    headers = {"Authorization": f"Bearer {token}"}
    t0 = time.time()
    while True:
        r = requests.get(f"{API}/status/{task_id}", headers=headers, allow_redirects=False)
        if r.status_code == 303:
            loc = r.headers.get("Location", "")
            return normalize_bundle_url(loc or f"/api/bundle/{task_id}", task_id)
        if r.status_code != 200:
            raise SystemExit(f"Status check failed ({r.status_code}): {r.text}")
        info = r.json()
        state = (info.get("status") or info.get("state") or "").lower()
        pct = info.get("progress") or info.get("percent") or info.get("summary") or ""
        details = info.get("details")
        print(f"Status: {state or pct} {pct if state else ''}".strip())
        if state == "done":
            return f"{API}/bundle/{task_id}"
        if (time.time() - t0) > max_wait_min * 60:
            raise SystemExit("Timed out waiting for AppEEARS task to finish.")
        time.sleep(sleep_s)

def list_bundle_files(token: str, bundle_url: str):
    headers = {"Authorization": f"Bearer {token}"}
    url = normalize_bundle_url(bundle_url)
    r = requests.get(url, headers=headers)
    r.raise_for_status()
    data = r.json()
    return data.get("task_id") or data.get("taskID"), data.get("files") or []

def download_bundle_file(token: str, task_id: str, file_id: str, out_dir: str):
    headers = {"Authorization": f"Bearer {token}"}
    url = f"{API}/bundle/{task_id}/{file_id}"
    r = requests.get(url, headers=headers, stream=True)
    r.raise_for_status()
    cd = r.headers.get("Content-Disposition", "")
    if "filename=" in cd:
        fname = cd.split("filename=")[-1].strip('" ')
    else:
        ext = ".tif" if "tif" in (r.headers.get("Content-Type", "")).lower() else ".zip"
        fname = f"{file_id}{ext}"
    out_path = os.path.join(out_dir, fname)
    with open(out_path, "wb") as f, tqdm(desc=f"Downloading {fname}", unit="B", unit_scale=True) as pbar:
        for chunk in r.iter_content(chunk_size=2**20):
            if chunk:
                f.write(chunk)
                pbar.update(len(chunk))
    return out_path

def unzip_if_zip(path: str, out_dir: str):
    if path.lower().endswith(".zip") and zipfile.is_zipfile(path):
        with zipfile.ZipFile(path, "r") as z:
            z.extractall(out_dir)
        print(f"Unzipped → {out_dir}")

def main():
    print("Earthdata Login required for AppEEARS.")
    user = input("Username: ").strip()
    pwd = getpass.getpass("Password: ")

    token = login(user, pwd)
    print("✔ Logged in to AppEEARS.")

    product_id, layer_name = get_ndvi_layer_name()
    print(f"Using layer: {product_id} / {layer_name}")

    task_name = f"Iowa_MOD13Q1_NDVI_{START.isoformat()}_{END.isoformat()}"
    print("Submitting area task …")
    task_id = submit_area_task(token, task_name, product_id, layer_name, START, END, BBOX)
    print(f"Task ID: {task_id}")

    print("Polling status (this can take several minutes)…")
    bundle_url = poll_until_done(token, task_id)
    print(f"✔ Task finished. Bundle: {bundle_url}")

    print("Listing bundle files …")
    task_id_from_bundle, files = list_bundle_files(token, bundle_url)
    if not files:
        print("No files in bundle.")
        return

    print(f"{len(files)} file(s) available. Downloading to {OUT_DIR} …")
    for f in files:
        file_id = f.get("file_id") or f.get("fileID") or f.get("id")
        if not file_id:
            # Last-resort direct URL entry
            url = f.get("url")
            if url:
                r = requests.get(url, stream=True)
                r.raise_for_status()
                name = os.path.basename(url.split("?")[0])
                out_path = os.path.join(OUT_DIR, name)
                with open(out_path, "wb") as h, tqdm(desc=f"Downloading {name}", unit="B", unit_scale=True) as pbar:
                    for chunk in r.iter_content(chunk_size=2**20):
                        if chunk:
                            h.write(chunk)
                            pbar.update(len(chunk))
                unzip_if_zip(out_path, OUT_DIR)
            continue
        out_path = download_bundle_file(token, task_id_from_bundle, file_id, OUT_DIR)
        unzip_if_zip(out_path, OUT_DIR)

    print("\nDone. Check:", os.path.abspath(OUT_DIR))
    print("You should see NDVI GeoTIFF(s) and a manifest JSON.")

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Iowa cornfields NDVI time series from AppEEARS MOD13Q1 GeoTIFFs

- Scans an AppEEARS download folder for NDVI GeoTIFFs
- Uses a cornfields vector (Shapefile/GeoJSON/etc.) to mask NDVI to corn only
- Aggregates mean NDVI per composite date
- (Optional) also outputs per-field means if an ID column exists

Outputs:
  - ndvi_iowa_corn_timeseries.csv
  - ndvi_iowa_corn_timeseries_by_field.csv  (optional, if PER_FIELD=True)

Dependencies (conda recommended):
  conda install -c conda-forge rasterio geopandas shapely pandas numpy tqdm fiona

Tested with:
  Python 3.10+, rasterio 1.3.x, geopandas 0.14.x
"""

import os
import re
import json
import glob
from datetime import datetime, timedelta
from typing import Dict, Tuple, Optional, List

import numpy as np
import pandas as pd
from tqdm import tqdm

import rasterio
from rasterio.mask import mask
from rasterio.enums import Resampling
import geopandas as gpd
from shapely.geometry import mapping

# -------------------- USER SETTINGS --------------------
APPEEARS_DIR = "appeears_ndvi_iowa"   # folder you printed: .../aashna_work/appeears_ndvi_iowa
CORN_VECTOR  = "cornfields_iowa.shp"  # <-- change to your cornfields file (Shapefile/GeoJSON/etc.)
OUT_CSV_ALL  = "ndvi_iowa_corn_timeseries.csv"
OUT_CSV_BYID = "ndvi_iowa_corn_timeseries_by_field.csv"

PER_FIELD = False  # set True to also compute per-field means (requires an ID column)
FIELD_ID_CANDIDATES = ["field_id", "Field_ID", "FIELD_ID", "id", "ID", "fid", "FID", "OBJECTID", "OBJECTID_1"]
# -------------------------------------------------------


def find_manifest_json(folder: str) -> Optional[str]:
    """
    Try to find an AppEEARS manifest/bundle JSON in the folder.
    We expect something with a 'files' list containing 'file_name' and dates.
    """
    for p in glob.glob(os.path.join(folder, "*.json")):
        try:
            with open(p, "r") as f:
                j = json.load(f)
            if isinstance(j, dict) and "files" in j and isinstance(j["files"], list):
                # looks like a bundle listing
                return p
            # Some manifests are nested under a 'files' key that contains dicts with file_name
            if isinstance(j, dict) and any(k in j for k in ("file_name", "fileName")):
                return p
        except Exception:
            pass
    return None


def parse_dates_from_manifest(manifest_path: str) -> Dict[str, Tuple[Optional[str], Optional[str]]]:
    """
    Returns mapping: basename.tif -> (start_date_iso, end_date_iso)
    Dates returned as ISO 'YYYY-MM-DD' if found, else None.
    """
    with open(manifest_path, "r") as f:
        j = json.load(f)

    mapping_dict: Dict[str, Tuple[Optional[str], Optional[str]]] = {}

    def norm_date(s: Optional[str]) -> Optional[str]:
        if not s:
            return None
        # Try common formats
        for fmt in ("%m-%d-%Y", "%Y-%m-%d", "%Y/%m/%d", "%m/%d/%Y", "%Y-%j"):
            try:
                dt = datetime.strptime(s, fmt)
                return dt.date().isoformat()
            except Exception:
                continue
        # If string contains ISO-looking date
        m = re.search(r"(\d{4})-(\d{2})-(\d{2})", s)
        if m:
            return f"{m.group(1)}-{m.group(2)}-{m.group(3)}"
        return None

    files = j.get("files") if isinstance(j, dict) else None
    if isinstance(files, list):
        for item in files:
            # common keys: file_name, fileName, start_date, end_date, startDate, endDate
            fname = item.get("file_name") or item.get("fileName") or item.get("name")
            if not fname:
                continue
            start = item.get("start_date") or item.get("startDate") or item.get("start")
            end   = item.get("end_date") or item.get("endDate") or item.get("end")
            start = norm_date(start)
            end   = norm_date(end)
            mapping_dict[os.path.basename(fname)] = (start, end)

    return mapping_dict


def parse_dates_from_filename(fname: str) -> Tuple[Optional[str], Optional[str]]:
    """
    Fallback: extract dates from filename if manifest isn't available.

    Heuristics:
      - Look for YYYY-MM-DD patterns (take first as start, last as end).
      - Else look for MODIS DOY 'Ayyyyddd' and convert to calendar date (use as end-date).
    """
    base = os.path.basename(fname)

    # 1) YYYY-MM-DD patterns
    matches = re.findall(r"(\d{4}-\d{2}-\d{2})", base)
    if matches:
        if len(matches) == 1:
            return matches[0], matches[0]
        return matches[0], matches[-1]

    # 2) Ayyyyddd (e.g., A2024185)
    m = re.search(r"A(\d{7})", base)
    if m:
        y = int(m.group(1)[:4])
        doy = int(m.group(1)[4:])
        dt = datetime.strptime(f"{y}-{doy:03d}", "%Y-%j").date().isoformat()
        return (None, dt)

    return (None, None)


def pick_id_column(gdf: gpd.GeoDataFrame) -> Optional[str]:
    for c in FIELD_ID_CANDIDATES:
        if c in gdf.columns:
            return c
    # if nothing obvious, try the index name or create one
    return None


def reproject_gdf(gdf: gpd.GeoDataFrame, crs_target) -> gpd.GeoDataFrame:
    if gdf.crs is None:
        # Assume it's in WGS84 if unspecified; adjust if your file has a known CRS
        gdf = gdf.set_crs("EPSG:4326", allow_override=True)
    if str(gdf.crs) != str(crs_target):
        gdf = gdf.to_crs(crs_target)
    return gdf


def mean_ndvi_for_geoms(src: rasterio.io.DatasetReader, geoms) -> Tuple[float, int]:
    """
    Mask src by geoms, compute mean NDVI ignoring NaNs/nodata.
    Returns (mean, count_of_valid_pixels)
    """
    # filled=False => returns a masked array; easier to do mean ignoring masked
    data, _ = mask(src, geoms, crop=True, nodata=src.nodata, filled=False)
    band = data[0]  # 1-band NDVI
    # Replace absurd values with masked
    # If values are already in [-1,1], fine; otherwise NaNs will be masked by 'mask'
    valid = band.compressed()  # 1-D array of unmasked
    if valid.size == 0:
        return (float("nan"), 0)
    return (float(valid.mean()), int(valid.size))


def process_all():
    # 1) Collect GeoTIFFs
    tifs = sorted([p for p in glob.glob(os.path.join(APPEEARS_DIR, "*.tif")) if os.path.isfile(p)])
    if not tifs:
        raise SystemExit(f"No .tif files found in {APPEEARS_DIR}")

    # 2) Map files to dates via manifest (preferred), else filename heuristic
    manifest = find_manifest_json(APPEEARS_DIR)
    date_map: Dict[str, Tuple[Optional[str], Optional[str]]] = {}
    if manifest:
        try:
            date_map = parse_dates_from_manifest(manifest)
            print(f"Using manifest to map dates: {os.path.basename(manifest)}")
        except Exception as e:
            print(f"Manifest parse failed ({e}); will fall back to filename parsing.")

    # 3) Load corn polygons
    if not os.path.exists(CORN_VECTOR):
        raise SystemExit(f"Corn polygons not found: {CORN_VECTOR}")
    gdf = gpd.read_file(CORN_VECTOR)
    if gdf.empty:
        raise SystemExit("Corn polygons file is empty.")
    id_col = pick_id_column(gdf) if PER_FIELD else None
    if PER_FIELD and id_col is None:
        print("PER_FIELD is True but no ID column found; will create sequential IDs.")
        gdf["__auto_id__"] = range(1, len(gdf) + 1)
        id_col = "__auto_id__"

    # 4) Iterate rasters and compute means
    rows_all: List[Dict] = []
    per_field_rows: List[Dict] = []

    for tif in tqdm(tifs, desc="Processing GeoTIFFs"):
        with rasterio.open(tif) as src:
            # Reproject polygons to raster CRS once per file
            gg = reproject_gdf(gdf, src.crs)
            geoms = [mapping(geom) for geom in gg.geometry.values if geom is not None]

            # dates
            base = os.path.basename(tif)
            if base in date_map:
                start_iso, end_iso = date_map[base]
            else:
                start_iso, end_iso = parse_dates_from_filename(base)

            # mean across all corn polygons together (dissolve for speed)
            gg_union = gg.unary_union
            mean_all, n_pix = mean_ndvi_for_geoms(src, [mapping(gg_union)])
            rows_all.append({
                "file": base,
                "start_date": start_iso,
                "end_date": end_iso,
                "date": end_iso or start_iso,   # single date column (prefer end-date)
                "ndvi_mean": mean_all,
                "n_pixels": n_pix
            })

            # optional: per-field
            if PER_FIELD:
                for idx, row in gg.iterrows():
                    geom = row.geometry
                    if geom is None or geom.is_empty:
                        continue
                    m, n = mean_ndvi_for_geoms(src, [mapping(geom)])
                    per_field_rows.append({
                        "file": base,
                        "start_date": start_iso,
                        "end_date": end_iso,
                        "date": end_iso or start_iso,
                        "field_id": row[id_col],
                        "ndvi_mean": m,
                        "n_pixels": n
                    })

    # 5) Save CSVs
    df = pd.DataFrame(rows_all)
    # Sort by date if available
    if "date" in df and df["date"].notna().any():
        df = df.sort_values(["date", "file"], na_position="last")
    df.to_csv(OUT_CSV_ALL, index=False)
    print(f"Wrote {OUT_CSV_ALL} ({len(df)} rows)")

    if PER_FIELD and per_field_rows:
        dff = pd.DataFrame(per_field_rows)
        if "date" in dff and dff["date"].notna().any():
            dff = dff.sort_values(["field_id", "date", "file"], na_position="last")
        dff.to_csv(OUT_CSV_BYID, index=False)
        print(f"Wrote {OUT_CSV_BYID} ({len(dff)} rows)")


if __name__ == "__main__":
    process_all()
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
AppEEARS MODIS NDVI → Iowa corn-only time series (handles ZIP → GeoTIFF or NetCDF)

What it does
------------
1) Recursively unzip all bundles in APPEEARS_DIR
2) Find .tif/.tiff (preferred). If none found, find .nc and convert each to GeoTIFF
3) Read corn polygons (Iowa-only), union mask, compute mean NDVI per composite date
4) Auto-detect NDVI scaling (10000 -> 1.0)
5) Save ndvi_iowa_corn_timeseries.csv (and optional per-field CSV)

Edit the USER SETTINGS block for your corn polygons path (Shapefile/GeoJSON).
"""

import os
import re
import json
import glob
import zipfile
import tempfile
from datetime import datetime
from typing import Dict, Tuple, Optional, List

import numpy as np
import pandas as pd
from tqdm import tqdm

import rasterio
from rasterio.mask import mask
from rasterio.enums import Resampling
import geopandas as gpd
from shapely.geometry import mapping

# Optional imports for NetCDF conversion (loaded lazily)
xr = None
rio = None

# -------------------- USER SETTINGS --------------------
APPEEARS_DIR = "/Users/sshome/Desktop/backup_Jan3_2025/aashna_work/appeears_ndvi_iowa"  # your AppEEARS folder
CORN_VECTOR  = "cornfields_iowa.shp"   # <-- path to your corn polygons (Iowa-only)

# Optional: also write per-field means
PER_FIELD    = False
OUT_CSV_ALL  = "ndvi_iowa_corn_timeseries.csv"
OUT_CSV_BYID = "ndvi_iowa_corn_timeseries_by_field.csv"
FIELD_ID_CANDIDATES = ["field_id","Field_ID","FIELD_ID","id","ID","fid","FID","OBJECTID","OBJECTID_1"]
# -------------------------------------------------------

FILL_VALUES = {-3000}
VALID_RANGE = (-2000, 10000)      # pre-scale range
AUTO_SCALE_THRESHOLD = 2.0        # if p95>2, scale by 1e-4

# -------------------- File helpers --------------------
def unzip_all(root: str) -> None:
    """Recursively unzip any .zip files under root."""
    zips = [p for p in glob.glob(os.path.join(root, "**", "*.zip"), recursive=True)]
    if not zips:
        return
    for z in zips:
        try:
            if zipfile.is_zipfile(z):
                out_dir = os.path.dirname(z)
                with zipfile.ZipFile(z, "r") as zz:
                    zz.extractall(out_dir)
                print(f"Unzipped: {os.path.relpath(z, root)}")
        except Exception as e:
            print(f"Warning: could not unzip {z}: {e}")

def find_all_tifs(root: str) -> List[str]:
    pats = ["**/*.tif", "**/*.tiff", "*.tif", "*.tiff"]
    found = []
    for p in pats:
        found += glob.glob(os.path.join(root, p), recursive=True)
    return sorted(list({os.path.abspath(p) for p in found if os.path.isfile(p)}))

def find_all_netcdf(root: str) -> List[str]:
    pats = ["**/*.nc", "*.nc", "**/*.nc4", "*.nc4"]
    found = []
    for p in pats:
        found += glob.glob(os.path.join(root, p), recursive=True)
    return sorted(list({os.path.abspath(p) for p in found if os.path.isfile(p)}))

def find_all_manifests(root: str) -> List[str]:
    return sorted(glob.glob(os.path.join(root, "**", "*.json"), recursive=True))

# -------------------- Date parsing --------------------
def parse_dates_from_manifest_files(json_paths: List[str]) -> Dict[str, Tuple[Optional[str], Optional[str]]]:
    """
    Map: basename -> (start_iso, end_iso) aggregated from any AppEEARS bundle JSONs.
    """
    def norm_date(s: Optional[str]) -> Optional[str]:
        if not s: return None
        fmts = ("%m-%d-%Y","%Y-%m-%d","%Y/%m/%d","%m/%d/%Y","%Y-%j")
        for fmt in fmts:
            try:
                return datetime.strptime(s, fmt).date().isoformat()
            except Exception:
                pass
        m = re.search(r"(\d{4})-(\d{2})-(\d{2})", s)
        return f"{m.group(1)}-{m.group(2)}-{m.group(3)}" if m else None

    out: Dict[str, Tuple[Optional[str], Optional[str]]] = {}
    for jp in json_paths:
        try:
            with open(jp, "r") as f:
                j = json.load(f)
        except Exception:
            continue
        files = j.get("files") if isinstance(j, dict) else None
        if isinstance(files, list):
            for it in files:
                fname = it.get("file_name") or it.get("fileName") or it.get("name")
                if not fname: continue
                start = it.get("start_date") or it.get("startDate") or it.get("start")
                end   = it.get("end_date")   or it.get("endDate")   or it.get("end")
                out[os.path.basename(fname)] = (norm_date(start), norm_date(end))
    return out

def dates_from_filename(path: str) -> Tuple[Optional[str], Optional[str]]:
    b = os.path.basename(path)
    m = re.findall(r"(\d{4}-\d{2}-\d{2})", b)
    if m:
        return (m[0], m[-1]) if len(m) > 1 else (m[0], m[0])
    m2 = re.search(r"A(\d{7})", b)
    if m2:
        y = int(m2.group(1)[:4]); doy = int(m2.group(1)[4:])
        dt = datetime.strptime(f"{y}-{doy:03d}", "%Y-%j").date().isoformat()
        return (None, dt)
    return (None, None)

# -------------------- Geo helpers --------------------
def reproject_gdf(gdf: gpd.GeoDataFrame, crs_target) -> gpd.GeoDataFrame:
    if gdf.crs is None:
        gdf = gdf.set_crs("EPSG:4326", allow_override=True)
    if str(gdf.crs) != str(crs_target):
        gdf = gdf.to_crs(crs_target)
    return gdf

def pick_id_column(gdf: gpd.GeoDataFrame) -> Optional[str]:
    for c in FIELD_ID_CANDIDATES:
        if c in gdf.columns:
            return c
    return None

# -------------------- NDVI read/scale --------------------
def read_ndvi_array(src: rasterio.io.DatasetReader, geoms) -> np.ndarray:
    """Mask by geoms, return float array with NaNs; auto-scale if needed."""
    data, _ = mask(src, geoms, crop=True, nodata=src.nodata, filled=False)
    arr = data[0].astype(np.float32)
    invalid = np.isin(arr, list(FILL_VALUES)) | (arr < VALID_RANGE[0]) | (arr > VALID_RANGE[1])
    arr = np.where(invalid, np.nan, arr)
    sample = arr[~np.isnan(arr)]
    if sample.size > 0:
        p95 = float(np.percentile(sample, 95))
        if p95 > AUTO_SCALE_THRESHOLD:
            arr = arr * 1e-4
    return arr

# -------------------- NetCDF → GeoTIFF --------------------
def ensure_imports_for_nc():
    global xr, rio
    if xr is None or rio is None:
        import xarray as _xr
        import rioxarray as _rio
        xr, rio = _xr, _rio

def guess_ndvi_variable(ds) -> Optional[str]:
    # Prefer var containing 'ndvi' (case-insensitive) and not 'qa'
    candidates = [v for v in ds.data_vars if "ndvi" in v.lower()]
    for v in candidates:
        if "qa" not in v.lower():
            return v
    # fallback: first var
    return candidates[0] if candidates else None

def nc_to_geotiff(nc_path: str, out_dir: str) -> Optional[str]:
    """
    Convert a single-band NDVI NetCDF to GeoTIFF with CRS & transform.
    Returns output path or None if conversion fails.
    """
    try:
        ensure_imports_for_nc()
        ds = xr.open_dataset(nc_path, decode_times=False)
        var = guess_ndvi_variable(ds)
        if var is None:
            # try any numeric var
            numeric_vars = [v for v in ds.data_vars if ds[v].dtype.kind in ("i","u","f")]
            if not numeric_vars:
                print(f"Skip (no numeric var): {os.path.basename(nc_path)}")
                return None
            var = numeric_vars[0]
        da = ds[var]

        # If dimensions are named lat/lon (common when projection='geographic')
        # otherwise, rioxarray will try to infer from attributes
        if set(["lat","latitude"]).intersection(da.coords) and set(["lon","longitude"]).intersection(da.coords):
            # normalize names
            if "latitude" in da.coords: da = da.rename({"latitude":"lat"})
            if "longitude" in da.coords: da = da.rename({"longitude":"lon"})
            da = da.rio.set_spatial_dims(x_dim="lon", y_dim="lat")
            da = da.rio.write_crs("EPSG:4326", inplace=True)
        else:
            # try to rely on CF metadata (grid_mapping) if present
            if not da.rio.crs:
                da = da.rio.write_crs("EPSG:4326", inplace=True)

        # squeeze band dimension if present (AppEEARS may use time/band=1)
        for dim in list(da.dims):
            if da.sizes[dim] == 1:
                da = da.squeeze(dim)

        base = os.path.basename(nc_path)
        out_name = re.sub(r"\.nc4?$", ".tif", base, flags=re.IGNORECASE)
        out_path = os.path.join(out_dir, out_name)
        da.rio.to_raster(out_path)
        ds.close()
        return out_path
    except Exception as e:
        print(f"NC→TIF failed for {os.path.basename(nc_path)}: {e}")
        return None

def convert_all_nc_to_tif(nc_paths: List[str], out_dir: str) -> List[str]:
    os.makedirs(out_dir, exist_ok=True)
    out_tifs = []
    for p in tqdm(nc_paths, desc="Converting NetCDF → GeoTIFF"):
        tif = nc_to_geotiff(p, out_dir)
        if tif and os.path.isfile(tif):
            out_tifs.append(os.path.abspath(tif))
    return out_tifs

# -------------------- Main --------------------
def main():
    if not os.path.isdir(APPEEARS_DIR):
        raise SystemExit(f"Folder not found: {APPEEARS_DIR}")

    # 1) unzip bundles
    unzip_all(APPEEARS_DIR)

    # 2) find GeoTIFFs
    tifs = find_all_tifs(APPEEARS_DIR)

    # 3) if no GeoTIFFs, try NetCDF and convert
    if not tifs:
        ncs = find_all_netcdf(APPEEARS_DIR)
        if not ncs:
            # Help: show a few entries
            entries = sorted(glob.glob(os.path.join(APPEEARS_DIR, "**", "*"), recursive=True))
            print(f"Found {len(entries)} entries; examples:")
            for p in entries[:15]:
                print("  ", os.path.relpath(p, APPEEARS_DIR))
            raise SystemExit("No .tif or .nc found. Re-run AppEEARS with output.format=geotiff or netcdf4.")
        # Convert NC to TIF into a temp subfolder
        conv_dir = os.path.join(APPEEARS_DIR, "_converted_tifs")
        tifs = convert_all_nc_to_tif(ncs, conv_dir)

    if not tifs:
        raise SystemExit("Still no GeoTIFFs after NetCDF conversion.")

    print(f"Using {len(tifs)} GeoTIFF(s). Example files:")
    for s in tifs[:3]:
        print("  -", os.path.basename(s))

    # 4) read manifest(s) for dates
    manifests = find_all_manifests(APPEEARS_DIR)
    date_map = parse_dates_from_manifest_files(manifests) if manifests else {}

    # 5) load corn polygons
    if not os.path.exists(CORN_VECTOR):
        raise SystemExit(f"Corn polygons not found: {CORN_VECTOR}")
    gdf = gpd.read_file(CORN_VECTOR)
    if gdf.empty:
        raise SystemExit("Corn polygons file is empty.")

    # per-field option
    id_col = None
    if PER_FIELD:
        for c in FIELD_ID_CANDIDATES:
            if c in gdf.columns:
                id_col = c
                break
        if id_col is None:
            gdf["__auto_id__"] = range(1, len(gdf) + 1)
            id_col = "__auto_id__"

    # 6) process rasters
    rows_all = []
    rows_by_field = []

    for tif in tqdm(tifs, desc="Processing NDVI"):
        with rasterio.open(tif) as src:
            gg = reproject_gdf(gdf, src.crs)
            union_geom = [mapping(gg.unary_union)]

            base = os.path.basename(tif)
            if base in date_map:
                start_iso, end_iso = date_map[base]
            else:
                start_iso, end_iso = dates_from_filename(base)
            date_iso = end_iso or start_iso

            arr = read_ndvi_array(src, union_geom)
            mean_all = float(np.nanmean(arr)) if np.isfinite(arr).any() else float("nan")
            n_valid = int(np.isfinite(arr).sum())

            rows_all.append({
                "file": base, "start_date": start_iso, "end_date": end_iso,
                "date": date_iso, "ndvi_mean": mean_all, "n_pixels": n_valid
            })

            if PER_FIELD:
                for _, row in gg.iterrows():
                    geom = row.geometry
                    if geom is None or geom.is_empty:
                        continue
                    arr_f = read_ndvi_array(src, [mapping(geom)])
                    mean_f = float(np.nanmean(arr_f)) if np.isfinite(arr_f).any() else float("nan")
                    n_f = int(np.isfinite(arr_f).sum())
                    rows_by_field.append({
                        "file": base, "start_date": start_iso, "end_date": end_iso,
                        "date": date_iso, "field_id": row[id_col], "ndvi_mean": mean_f, "n_pixels": n_f
                    })

    # 7) write CSVs
    df = pd.DataFrame(rows_all)
    if "date" in df and df["date"].notna().any():
        df = df.sort_values(["date", "file"], na_position="last")
    df.to_csv(OUT_CSV_ALL, index=False)
    print(f"Wrote {OUT_CSV_ALL} ({len(df)} rows)")

    if PER_FIELD and rows_by_field:
        dff = pd.DataFrame(rows_by_field)
        if "date" in dff and dff["date"].notna().any():
            dff = dff.sort_values(["field_id","date","file"], na_position="last")
        dff.to_csv(OUT_CSV_BYID, index=False)
        print(f"Wrote {OUT_CSV_BYID} ({len(dff)} rows)")

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
AppEEARS MODIS NDVI → Iowa corn-only time series
- Unzips all bundles under APPEEARS_DIR
- Extracts only .tif/.tiff or .nc/.nc4 into APPEEARS_DIR/_unzipped
- If only NetCDF is present, converts to GeoTIFF (xarray+rioxarray)
- Masks with corn polygons and computes mean NDVI per composite date
- Auto-detects NDVI scaling (10000 -> 1.0)
- Writes ndvi_iowa_corn_timeseries.csv (and optional per-field CSV)
"""

import os
import re
import json
import glob
import zipfile
from datetime import datetime
from typing import Dict, Tuple, Optional, List

import numpy as np
import pandas as pd
from tqdm import tqdm

import rasterio
from rasterio.mask import mask
import geopandas as gpd
from shapely.geometry import mapping

# =============== USER SETTINGS ===============
APPEEARS_DIR = "/Users/sshome/Desktop/backup_Jan3_2025/aashna_work/appeears_ndvi_iowa"  # <- your folder
CORN_VECTOR  = "cornfields_iowa.shp"   # <- set to your Iowa corn polygons (Shapefile/GeoJSON/etc.)

PER_FIELD    = False                   # True -> also compute per-field means (needs an ID column)
OUT_CSV_ALL  = "ndvi_iowa_corn_timeseries.csv"
OUT_CSV_BYID = "ndvi_iowa_corn_timeseries_by_field.csv"
FIELD_ID_CANDIDATES = ["field_id","Field_ID","FIELD_ID","id","ID","fid","FID","OBJECTID","OBJECTID_1"]
# ============================================

FILL_VALUES = {-3000}
VALID_RANGE = (-2000, 10000)
AUTO_SCALE_THRESHOLD = 2.0  # if NDVI p95 > 2, assume 0..10000 and scale by 1e-4

UNZIP_DIR = os.path.join(APPEEARS_DIR, "_unzipped")
CONVERT_DIR = os.path.join(APPEEARS_DIR, "_converted_tifs")

def ensure_dirs():
    os.makedirs(UNZIP_DIR, exist_ok=True)
    os.makedirs(CONVERT_DIR, exist_ok=True)

def list_zip_preview(root: str, limit: int = 3):
    zips = sorted(glob.glob(os.path.join(root, "*.zip")))
    print(f"Found {len(zips)} zip(s). Previewing first {min(limit,len(zips))}:")
    for z in zips[:limit]:
        print(" -", os.path.basename(z))
        try:
            with zipfile.ZipFile(z, "r") as zz:
                names = zz.namelist()
                print("   contents (first 8):", names[:8])
        except Exception as e:
            print("   (cannot open zip):", e)

def unzip_extract_selected(root: str, out_dir: str) -> int:
    """
    Extract only NDVI data files from each zip into out_dir, preserving zip paths.
    Returns number of files extracted.
    """
    patterns = (".tif", ".tiff", ".nc", ".nc4")
    count = 0
    zips = sorted(glob.glob(os.path.join(root, "*.zip")))
    for z in tqdm(zips, desc="Unzipping bundles"):
        try:
            with zipfile.ZipFile(z, "r") as zz:
                for name in zz.namelist():
                    low = name.lower()
                    if low.endswith(patterns):
                        # Flatten long paths a bit but keep uniqueness
                        tgt = os.path.join(out_dir, os.path.basename(name))
                        # If collision, prefix with zip stem
                        if os.path.exists(tgt):
                            stem = os.path.splitext(os.path.basename(z))[0]
                            tgt = os.path.join(out_dir, f"{stem}__{os.path.basename(name)}")
                        with zz.open(name) as src, open(tgt, "wb") as dst:
                            dst.write(src.read())
                        count += 1
        except Exception as e:
            print(f"Warning: cannot unzip {os.path.basename(z)}: {e}")
    return count

def find_all_tifs(root: str) -> List[str]:
    pats = ["**/*.tif", "**/*.tiff", "*.tif", "*.tiff"]
    found = []
    for p in pats:
        found += glob.glob(os.path.join(root, p), recursive=True)
    return sorted(list({os.path.abspath(p) for p in found if os.path.isfile(p)}))

def find_all_netcdf(root: str) -> List[str]:
    pats = ["**/*.nc", "**/*.nc4", "*.nc", "*.nc4"]
    found = []
    for p in pats:
        found += glob.glob(os.path.join(root, p), recursive=True)
    return sorted(list({os.path.abspath(p) for p in found if os.path.isfile(p)}))

def find_all_manifests(root: str) -> List[str]:
    return sorted(glob.glob(os.path.join(root, "**", "*.json"), recursive=True))

def parse_dates_from_manifest_files(json_paths: List[str]) -> Dict[str, Tuple[Optional[str], Optional[str]]]:
    def norm_date(s: Optional[str]) -> Optional[str]:
        if not s: return None
        fmts = ("%m-%d-%Y","%Y-%m-%d","%Y/%m/%d","%m/%d/%Y","%Y-%j")
        for fmt in fmts:
            try:
                return datetime.strptime(s, fmt).date().isoformat()
            except Exception:
                pass
        m = re.search(r"(\d{4})-(\d{2})-(\d{2})", s)
        return f"{m.group(1)}-{m.group(2)}-{m.group(3)}" if m else None
    out: Dict[str, Tuple[Optional[str], Optional[str]]] = {}
    for jp in json_paths:
        try:
            with open(jp, "r") as f: j = json.load(f)
        except Exception:
            continue
        files = j.get("files") if isinstance(j, dict) else None
        if isinstance(files, list):
            for it in files:
                fname = it.get("file_name") or it.get("fileName") or it.get("name")
                if not fname: continue
                start = it.get("start_date") or it.get("startDate") or it.get("start")
                end   = it.get("end_date")   or it.get("endDate")   or it.get("end")
                out[os.path.basename(fname)] = (norm_date(start), norm_date(end))
    return out

def dates_from_filename(path: str) -> Tuple[Optional[str], Optional[str]]:
    b = os.path.basename(path)
    m = re.findall(r"(\d{4}-\d{2}-\d{2})", b)
    if m: return (m[0], m[-1]) if len(m) > 1 else (m[0], m[0])
    m2 = re.search(r"A(\d{7})", b)
    if m2:
        y = int(m2.group(1)[:4]); doy = int(m2.group(1)[4:])
        dt = datetime.strptime(f"{y}-{doy:03d}", "%Y-%j").date().isoformat()
        return (None, dt)
    return (None, None)

def reproject_gdf(gdf: gpd.GeoDataFrame, crs_target) -> gpd.GeoDataFrame:
    if gdf.crs is None:
        gdf = gdf.set_crs("EPSG:4326", allow_override=True)
    if str(gdf.crs) != str(crs_target):
        gdf = gdf.to_crs(crs_target)
    return gdf

def pick_id_column(gdf: gpd.GeoDataFrame) -> Optional[str]:
    for c in FIELD_ID_CANDIDATES:
        if c in gdf.columns: return c
    return None

def read_ndvi_array(src: rasterio.io.DatasetReader, geoms) -> np.ndarray:
    data, _ = mask(src, geoms, crop=True, nodata=src.nodata, filled=False)
    arr = data[0].astype(np.float32)
    invalid = np.isin(arr, list(FILL_VALUES)) | (arr < VALID_RANGE[0]) | (arr > VALID_RANGE[1])
    arr = np.where(invalid, np.nan, arr)
    sample = arr[~np.isnan(arr)]
    if sample.size > 0:
        p95 = float(np.percentile(sample, 95))
        if p95 > AUTO_SCALE_THRESHOLD:
            arr = arr * 1e-4
    return arr

# ---------- NetCDF → GeoTIFF ----------
xr = None
rio = None
def ensure_nc_imports():
    global xr, rio
    if xr is None or rio is None:
        import xarray as _xr
        import rioxarray as _rio
        xr, rio = _xr, _rio

def guess_ndvi_variable(ds) -> Optional[str]:
    cand = [v for v in ds.data_vars if "ndvi" in v.lower()]
    cand = [v for v in cand if "qa" not in v.lower()] or cand
    return cand[0] if cand else None

def nc_to_geotiff(nc_path: str, out_dir: str) -> Optional[str]:
    try:
        ensure_nc_imports()
        ds = xr.open_dataset(nc_path, decode_times=False)
        var = guess_ndvi_variable(ds)
        if var is None:
            nv = [v for v in ds.data_vars if ds[v].dtype.kind in ("i","u","f")]
            if not nv: 
                print(f"Skip (no numeric var): {os.path.basename(nc_path)}"); return None
            var = nv[0]
        da = ds[var]

        # Normalize coords if lat/lon
        if set(["lat","latitude"]).intersection(da.coords) and set(["lon","longitude"]).intersection(da.coords):
            if "latitude" in da.coords: da = da.rename({"latitude":"lat"})
            if "longitude" in da.coords: da = da.rename({"longitude":"lon"})
            da = da.rio.set_spatial_dims(x_dim="lon", y_dim="lat")
            da = da.rio.write_crs("EPSG:4326", inplace=True)
        else:
            if not da.rio.crs:
                da = da.rio.write_crs("EPSG:4326", inplace=True)

        # Squeeze singleton dims
        for dim in list(da.dims):
            if da.sizes[dim] == 1:
                da = da.squeeze(dim)

        base = os.path.basename(nc_path)
        out_name = re.sub(r"\.nc4?$", ".tif", base, flags=re.IGNORECASE)
        out_path = os.path.join(out_dir, out_name)
        da.rio.to_raster(out_path)
        ds.close()
        return out_path
    except Exception as e:
        print(f"NC→TIF failed for {os.path.basename(nc_path)}: {e}")
        return None

def convert_all_nc_to_tif(nc_paths: List[str], out_dir: str) -> List[str]:
    os.makedirs(out_dir, exist_ok=True)
    out = []
    for p in tqdm(nc_paths, desc="Converting NetCDF → GeoTIFF"):
        op = nc_to_geotiff(p, out_dir)
        if op and os.path.isfile(op): out.append(os.path.abspath(op))
    return out

def main():
    if not os.path.isdir(APPEEARS_DIR):
        raise SystemExit(f"Folder not found: {APPEEARS_DIR}")
    ensure_dirs()

    # Preview a couple zips to see what's inside
    list_zip_preview(APPEEARS_DIR, limit=3)

    # Extract only relevant data files from each zip
    extracted = unzip_extract_selected(APPEEARS_DIR, UNZIP_DIR)
    print(f"Extracted {extracted} data file(s) (.tif/.tiff/.nc/.nc4) into {UNZIP_DIR}")

    # Gather GeoTIFFs first
    tifs = find_all_tifs(UNZIP_DIR)
    # If none, try NetCDF and convert to GeoTIFF
    if not tifs:
        ncs = find_all_netcdf(UNZIP_DIR)
        if not ncs:
            # Help: print some entries so we can see what's inside the zips
            entries = sorted(glob.glob(os.path.join(APPEEARS_DIR, "**", "*"), recursive=True))
            print(f"No .tif/.nc extracted. Example entries under APPEEARS_DIR:")
            for p in entries[:20]:
                print("  ", os.path.relpath(p, APPEEARS_DIR))
            raise SystemExit("No GeoTIFF or NetCDF found in bundles. Re-run AppEEARS with output.format geotiff or netcdf4.")
        print(f"Found {len(ncs)} NetCDF file(s), converting → GeoTIFF …")
        tifs = convert_all_nc_to_tif(ncs, CONVERT_DIR)

    if not tifs:
        raise SystemExit("No GeoTIFFs available after extraction/conversion.")

    print(f"Using {len(tifs)} GeoTIFF(s). Examples:")
    for s in tifs[:3]:
        print("  -", os.path.basename(s))

    # Date mapping from manifests (optional)
    manifests = find_all_manifests(APPEEARS_DIR)
    date_map = parse_dates_from_manifest_files(manifests) if manifests else {}

    # Load corn polygons
    if not os.path.exists(CORN_VECTOR):
        raise SystemExit(f"Corn polygons not found: {CORN_VECTOR}")
    gdf = gpd.read_file(CORN_VECTOR)
    if gdf.empty:
        raise SystemExit("Corn polygons file is empty.")

    id_col = None
    if PER_FIELD:
        id_col = pick_id_column(gdf)
        if id_col is None:
            gdf["__auto_id__"] = range(1, len(gdf)+1)
            id_col = "__auto_id__"

    # Process rasters
    rows_all: List[Dict] = []
    rows_by_field: List[Dict] = []

    for tif in tqdm(tifs, desc="Processing NDVI"):
        with rasterio.open(tif) as src:
            gg = reproject_gdf(gdf, src.crs)
            union_geom = [mapping(gg.unary_union)]

            base = os.path.basename(tif)
            if base in date_map:
                start_iso, end_iso = date_map[base]
            else:
                start_iso, end_iso = dates_from_filename(base)
            date_iso = end_iso or start_iso

            arr = read_ndvi_array(src, union_geom)
            mean_all = float(np.nanmean(arr)) if np.isfinite(arr).any() else float("nan")
            n_valid = int(np.isfinite(arr).sum())

            rows_all.append({
                "file": base, "start_date": start_iso, "end_date": end_iso,
                "date": date_iso, "ndvi_mean": mean_all, "n_pixels": n_valid
            })

            if PER_FIELD:
                for _, row in gg.iterrows():
                    geom = row.geometry
                    if geom is None or geom.is_empty: continue
                    arr_f = read_ndvi_array(src, [mapping(geom)])
                    mean_f = float(np.nanmean(arr_f)) if np.isfinite(arr_f).any() else float("nan")
                    n_f = int(np.isfinite(arr_f).sum())
                    rows_by_field.append({
                        "file": base, "start_date": start_iso, "end_date": end_iso,
                        "date": date_iso, "field_id": row[id_col], "ndvi_mean": mean_f, "n_pixels": n_f
                    })

    # Write CSVs
    df = pd.DataFrame(rows_all)
    if "date" in df and df["date"].notna().any():
        df = df.sort_values(["date","file"], na_position="last")
    df.to_csv(OUT_CSV_ALL, index=False)
    print(f"Wrote {OUT_CSV_ALL} ({len(df)} rows)")

    if PER_FIELD and rows_by_field:
        dff = pd.DataFrame(rows_by_field)
        if "date" in dff and dff["date"].notna().any():
            dff = dff.sort_values(["field_id","date","file"], na_position="last")
        dff.to_csv(OUT_CSV_BYID, index=False)
        print(f"Wrote {OUT_CSV_BYID} ({len(dff)} rows)")

if __name__ == "__main__":
    main()
```

```python
# %% REAL MODE (no demo). End-to-end extraction → seasonal sequences → LSTM training + diagnostics
import os, re, glob, math, sys, warnings
from datetime import datetime, timedelta
from typing import Optional, Tuple, Dict, Any, List

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Geospatial
import rasterio
from rasterio.enums import Resampling
from rasterio.warp import reproject
from rasterio.features import rasterize
import geopandas as gpd
from affine import Affine

import torch
from torch import nn

# ===================== CONFIG — EDIT THESE =====================
MOSAICS_DIR = "/ABS/PATH/TO/your/mosaics"   # folder containing NDVI mosaics
PATTERN     = "NDVI_*.tif"                  # adjust if your files differ
CDL_PATH    = "/ABS/PATH/TO/CDL_IA_2023_30m.tif"   # set this OR POLY_PATH (but not both)
POLY_PATH   = None                          # e.g., "/ABS/PATH/TO/fields.shp" (optionally with 'field_id')

# Optional: yields CSV with columns: field_id, year, yield (or change YIELD_COL below)
YIELD_CSV   = None                          # e.g., "/ABS/PATH/yields.csv"
YIELD_COL   = "yield"

# NDVI scale: set to 0.0001 if your tiles are integers 0..10000; otherwise None to auto-read
SCALE: Optional[float] = None
NO_CLIP = False
CLIP_RANGE = (-1.0, 1.0)

# CDL / rasterization
CDL_CLASS   = 1
ALL_TOUCHED = False

# Season window; if empty season detected, we can auto-infer (see AUTO_SEASON_FROM_DATA)
SEASON_DOY_LOW  = 100
SEASON_DOY_HIGH = 300
AUTO_SEASON_FROM_DATA = True

# Resampling for sequences
N_STEPS = 32
MIN_POINTS_PER_SEASON = 3

# Training config
TEST_SIZE = 0.2
VAL_SIZE  = 0.2
BATCH_SIZE= 32
EPOCHS    = 50
LR        = 1e-3
HIDDEN_DIM= 64
NUM_LAYERS= 1
PATIENCE  = 6
SEED      = 42

# Outputs
OUT_DIR   = "crop_yield_results"
OUT_TS_CSV   = os.path.join(OUT_DIR, "ndvi_timeseries_per_field.csv")
OUT_SEAS_CSV = os.path.join(OUT_DIR, "ndvi_seasonal_sequences.csv")
OUT_PRED_CSV = os.path.join(OUT_DIR, "yield_predictions.csv")
MODEL_PATH   = os.path.join(OUT_DIR, "lstm_yield_model.pt")
# ===============================================================

np.random.seed(SEED); torch.manual_seed(SEED)
os.makedirs(OUT_DIR, exist_ok=True)

def _fail(msg):
    raise SystemExit("❌ " + msg)

# --------- Robust date parsing from filename ----------
def parse_date_from_name(name: str) -> Optional[str]:
    """
    Try multiple patterns:
      - NDVI_YYYYMMDD.* or anything containing YYYYMMDD
      - MODIS 'A' year + DOY like A2020150  (YYYY DDD)
    Returns ISO date string 'YYYY-MM-DD' or None.
    """
    # exact NDVI_YYYYMMDD
    m = re.search(r"NDVI_(\d{8})", name, re.I)
    if m:
        return datetime.strptime(m.group(1), "%Y%m%d").date().isoformat()

    # any contiguous 8-digit date
    m = re.search(r"(\d{8})", name)
    if m:
        s = m.group(1)
        try:
            return datetime.strptime(s, "%Y%m%d").date().isoformat()
        except Exception:
            pass

    # MODIS AYYYYDDD
    m = re.search(r"A(\d{7})", name, re.I)
    if m:
        s = m.group(1)            # e.g., 2020150
        year = int(s[:4]); doy = int(s[4:])
        try:
            dt = datetime(year, 1, 1) + timedelta(days=doy-1)
            return dt.date().isoformat()
        except Exception:
            pass
    return None

def doy_from_iso(iso_date: str) -> int:
    return datetime.fromisoformat(iso_date).timetuple().tm_yday

def get_scale_and_offset(src, user_scale: Optional[float]) -> Tuple[float, float]:
    if user_scale is not None:
        return float(user_scale), 0.0
    scale = 1.0; offset = 0.0
    try:
        if getattr(src, "scales", None): 
            s = src.scales[0]
            if s not in (None, 1.0): scale = float(s)
        if getattr(src, "offsets", None):
            o = src.offsets[0]
            if o not in (None, 0.0): offset = float(o)
    except Exception: pass
    try:
        tags = src.tags()
        if "scale_factor" in tags: scale = float(tags["scale_factor"])
        if "add_offset"   in tags: offset = float(tags["add_offset"])
    except Exception: pass
    return scale, offset

def clean_ndvi(arr: np.ndarray, nodata_value, do_clip: bool, low: float, high: float) -> np.ndarray:
    if nodata_value is not None and not (isinstance(nodata_value, float) and math.isnan(nodata_value)):
        arr = np.where(arr == nodata_value, np.nan, arr)
    arr = np.where(arr < -1e19, np.nan, arr)
    if do_clip: arr = np.clip(arr, low, high)
    return arr

def build_mask_from_cdl(profile: Dict[str, Any], cdl_path: str, corn_code: int) -> np.ndarray:
    with rasterio.open(cdl_path) as src:
        dst = np.zeros((profile["height"], profile["width"]), dtype=np.uint16)
        reproject(
            source=rasterio.band(src, 1), destination=dst,
            src_transform=src.transform, src_crs=src.crs,
            dst_transform=profile["transform"], dst_crs=profile["crs"],
            resampling=Resampling.nearest, dst_nodata=0
        )
    return (dst == corn_code)

def build_field_id_raster(profile: Dict[str, Any], poly_path: str) -> np.ndarray:
    gdf = gpd.read_file(poly_path)
    if gdf.empty: _fail("Polygon file is empty.")
    if gdf.crs is None: _fail("Polygon file has no CRS defined.")
    id_col = None
    for c in ["field_id","Field_ID","field","FID","id","ID"]:
        if c in gdf.columns: id_col = c; break
    if id_col is None:
        gdf = gdf.reset_index(drop=False).rename(columns={"index":"field_id"})
        id_col = "field_id"
    gg = gdf.to_crs(profile["crs"])
    shapes = [(geom, int(fid)) for geom,fid in zip(gg.geometry, gg[id_col]) if geom is not None and not geom.is_empty]
    fid_arr = rasterize(
        shapes=shapes, out_shape=(profile["height"], profile["width"]),
        transform=profile["transform"], fill=0, all_touched=ALL_TOUCHED, dtype="int32"
    )
    return fid_arr

def sanity_checks():
    if not os.path.isdir(MOSAICS_DIR):
        _fail(f"MOSAICS_DIR not found: {MOSAICS_DIR}")
    tifs = sorted(glob.glob(os.path.join(MOSAICS_DIR, PATTERN)))
    if not tifs:
        _fail(f"No mosaics matching pattern {PATTERN!r} in {MOSAICS_DIR}")
    if (CDL_PATH is None) == (POLY_PATH is None):
        _fail("Provide exactly one of CDL_PATH or POLY_PATH.")
    if CDL_PATH and not os.path.isfile(CDL_PATH):
        _fail(f"CDL_PATH not found: {CDL_PATH}")
    if POLY_PATH and not os.path.exists(POLY_PATH):
        _fail(f"POLY_PATH not found: {POLY_PATH}")
    # Probe a sample
    with rasterio.open(tifs[0]) as src:
        dtype = src.dtypes[0]; nod = src.nodata
    print(f"✔ Found {len(tifs)} mosaics. Sample dtype={dtype}, nodata={nod}")
    return tifs

def extract_ndvi_timeseries_real(tifs: List[str]) -> pd.DataFrame:
    rows = []
    cache = None
    dates_missing = 0
    for tif in tifs:
        date_iso = parse_date_from_name(os.path.basename(tif))
        if not date_iso:
            dates_missing += 1
            continue
        with rasterio.open(tif) as src:
            arr = src.read(1, masked=False).astype(np.float32)
            scale, offset = get_scale_and_offset(src, SCALE)
            if scale != 1.0 or offset != 0.0:
                arr = arr * scale + offset
            arr = clean_ndvi(arr, src.nodata, not NO_CLIP, CLIP_RANGE[0], CLIP_RANGE[1])

            profile = {"height": src.height, "width": src.width, "crs": src.crs, "transform": src.transform}
            key = (str(src.crs), tuple(src.transform) if hasattr(src.transform, "__iter__") else src.transform, src.width, src.height)
            if cache is None or cache["key"] != key:
                if POLY_PATH:
                    fid = build_field_id_raster(profile, POLY_PATH)
                    cache = {"key": key, "mode":"poly", "fid": fid}
                else:
                    mask = build_mask_from_cdl(profile, CDL_PATH, CDL_CLASS)
                    cache = {"key": key, "mode":"cdl", "mask": mask}

            if cache["mode"] == "poly":
                fid = cache["fid"]
                valid = np.isfinite(arr) & (fid > 0)
                if valid.any():
                    s = pd.Series(arr[valid], index=fid[valid])
                    means = s.groupby(level=0).mean()
                    counts = s.groupby(level=0).size()
                    for f,v in means.items():
                        rows.append({
                            "date": date_iso, "year": int(date_iso[:4]), "doy": doy_from_iso(date_iso),
                            "field_id": int(f), "ndvi_mean": float(v), "n_pixels": int(counts.loc[f])
                        })
            else:
                mask = cache["mask"]
                valid = np.isfinite(arr) & mask
                ndvi_mean = float(np.nanmean(arr[valid])) if valid.any() else float("nan")
                n_pix = int(valid.sum())
                rows.append({
                    "date": date_iso, "year": int(date_iso[:4]), "doy": doy_from_iso(date_iso),
                    "field_id": 1, "ndvi_mean": ndvi_mean, "n_pixels": n_pix
                })
    if dates_missing:
        print(f"⚠️ Skipped {dates_missing} file(s) with unrecognized date pattern. Adjust PATTERN or parser if needed.")
    df = pd.DataFrame(rows).sort_values(["field_id","date"])
    return df

def infer_season(df: pd.DataFrame) -> Tuple[int,int]:
    ds = df["doy"].values
    if len(ds)==0: return SEASON_DOY_LOW, SEASON_DOY_HIGH
    lo = int(np.percentile(ds, 10)); hi = int(np.percentile(ds, 90))
    lo = max(1, lo); hi = min(365, hi)
    if hi - lo < 20:  # too narrow → fallback to config
        return SEASON_DOY_LOW, SEASON_DOY_HIGH
    return lo, hi

def build_season_sequences(df_long: pd.DataFrame, low:int, high:int, n_steps:int) -> pd.DataFrame:
    seq_rows = []
    for (fid, yr), g in df_long.groupby(["field_id","year"]):
        gg = g[(g["doy"]>=low) & (g["doy"]<=high)].sort_values("doy")
        gg = gg.dropna(subset=["ndvi_mean"])
        if len(gg) < MIN_POINTS_PER_SEASON:
            continue
        xs = gg["doy"].to_numpy(); ys = gg["ndvi_mean"].to_numpy()
        target = np.linspace(low, high, n_steps)
        ux, idx = np.unique(xs, return_index=True); uy = ys[idx]
        seq = np.repeat(uy[0], n_steps) if len(ux)==1 else np.interp(target, ux, uy)
        auc  = float(np.trapz(seq, target))  # ok in current NumPy; trapezoid also fine
        peak = float(np.nanmax(seq)); meanv = float(np.nanmean(seq))
        row = {"field_id": int(fid), "year": int(yr), "AUC": auc, "peak": peak, "mean_ndvi": meanv}
        for i,v in enumerate(seq): row[f"step_{i:02d}"]=float(v)
        seq_rows.append(row)
    return pd.DataFrame(seq_rows).sort_values(["field_id","year"])

class YieldLSTM(nn.Module):
    def __init__(self, input_dim=1, hidden_dim=HIDDEN_DIM, num_layers=NUM_LAYERS):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers=num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1)
    def forward(self, x):
        out,_ = self.lstm(x); return self.fc(out[:,-1,:])

def make_tensors(df: pd.DataFrame, step_cols: List[str]):
    X = df[step_cols].to_numpy(dtype=np.float32)
    x_mean = X.mean(axis=0, keepdims=True); x_std = X.std(axis=0, keepdims=True)+1e-8
    Xn = ((X - x_mean)/x_std)[:, :, None]
    y  = df[[YIELD_COL]].to_numpy(dtype=np.float32)
    return Xn, y, x_mean.astype(np.float32), x_std.astype(np.float32)

def split_df(df, test_size, val_size, seed):
    rng = np.random.default_rng(seed); idx = np.arange(len(df)); rng.shuffle(idx)
    n_test = max(1, int(len(df)*test_size)); n_val = max(1, int((len(df)-n_test)*val_size))
    test_idx = idx[:n_test]; val_idx = idx[n_test:n_test+n_val]; train_idx = idx[n_test+n_val:]
    return df.iloc[train_idx], df.iloc[val_idx], df.iloc[test_idx]

def train_lstm(X_train, y_train, X_val, y_val, device="cpu"):
    model = YieldLSTM().to(device)
    opt = torch.optim.Adam(model.parameters(), lr=LR)
    loss_fn = nn.MSELoss()
    def batches(X,y,bs):
        n=len(X); idx=np.arange(n)
        for i in range(0,n,bs): j=idx[i:i+bs]; yield X[j], y[j]
    best=float("inf"); best_state=None; patience=PATIENCE
    for ep in range(1,EPOCHS+1):
        model.train(); tl=[]
        for xb,yb in batches(X_train,y_train,BATCH_SIZE):
            xb=torch.from_numpy(xb).to(device); yb=torch.from_numpy(yb).to(device)
            opt.zero_grad(); pred=model(xb); loss=loss_fn(pred,yb); loss.backward(); opt.step(); tl.append(loss.item())
        model.eval()
        with torch.no_grad():
            xv=torch.from_numpy(X_val).to(device); yv=torch.from_numpy(y_val).to(device)
            vloss=loss_fn(model(xv), yv).item()
        print(f"Epoch {ep:03d}  train_loss={np.mean(tl):.4f}  val_loss={vloss:.4f}  (patience {patience})")
        if vloss < best - 1e-6:
            best=vloss; best_state={k:v.cpu().clone() for k,v in model.state_dict().items()}; patience=PATIENCE
        else:
            patience -= 1
            if patience==0: print("Early stopping."); break
    if best_state: model.load_state_dict(best_state)
    return model

def rmse(y,yhat): y=np.asarray(y); yhat=np.asarray(yhat); return float(np.sqrt(np.mean((y-yhat)**2)))
def r2(y,yhat):
    y=np.asarray(y); yhat=np.asarray(yhat)
    ss_res=np.sum((y-yhat)**2); ss_tot=np.sum((y-y.mean())**2)
    return float(1-ss_res/ss_tot) if ss_tot>0 else np.nan
def mae(y,yhat): y=np.asarray(y); yhat=np.asarray(yhat); return float(np.mean(np.abs(y-yhat)))

# ---------------- RUN (REAL ONLY) ----------------
tifs = sanity_checks()

# Quick mask overlap sanity (first mosaic)
with rasterio.open(tifs[0]) as src0:
    profile0 = {"height": src0.height, "width": src0.width, "crs": src0.crs, "transform": src0.transform}
    if POLY_PATH:
        fid0 = build_field_id_raster(profile0, POLY_PATH)
        frac = (fid0>0).sum() / float(fid0.size)
        print(f"✔ Field mask overlaps {frac*100:.2f}% of the mosaic grid.")
    else:
        mask0 = build_mask_from_cdl(profile0, CDL_PATH, CDL_CLASS)
        frac = mask0.sum() / float(mask0.size)
        print(f"✔ CDL corn mask covers {frac*100:.2f}% of the mosaic grid.")

# Extract per-date NDVI
df_long = extract_ndvi_timeseries_real(tifs)
if df_long.empty:
    _fail("No NDVI rows extracted. Check filename date pattern, masks, and SCALE.")
df_long.to_csv(OUT_TS_CSV, index=False)
print(f"✓ Wrote per-date NDVI → {OUT_TS_CSV}  ({len(df_long)} rows)")
print(df_long.head())

# Season window
low, high = SEASON_DOY_LOW, SEASON_DOY_HIGH
if AUTO_SEASON_FROM_DATA:
    auto_lo, auto_hi = infer_season(df_long)
    print(f"Season DOY window: configured [{SEASON_DOY_LOW},{SEASON_DOY_HIGH}], inferred [{auto_lo},{auto_hi}]")
    # use the overlap of configured & inferred to be safe
    low = max(SEASON_DOY_LOW, auto_lo); high = min(SEASON_DOY_HIGH, auto_hi)
    if high - low < 20:
        print("⚠️ Inferred window too narrow; falling back to configured window.")
        low, high = SEASON_DOY_LOW, SEASON_DOY_HIGH

# Build seasonal sequences
seq_df = build_season_sequences(df_long, low, high, N_STEPS)
if seq_df.empty:
    _fail("Seasonal sequences empty. Increase date coverage or relax season DOY window.")
seq_df.to_csv(OUT_SEAS_CSV, index=False)
print(f"✓ Wrote seasonal sequences → {OUT_SEAS_CSV}  ({len(seq_df)} rows)")
print(seq_df.head())

# Attach yields (real or simulated)
if YIELD_CSV and os.path.isfile(YIELD_CSV):
    ydf = pd.read_csv(YIELD_CSV)
    need = {"field_id","year",YIELD_COL}
    if not need.issubset(ydf.columns): _fail(f"Yield CSV must have columns {need}")
    data = seq_df.merge(ydf[list(need)], on=["field_id","year"], how="inner")
    if data.empty: _fail("No overlap between sequences and yields (check field_id/year alignment).")
    print(f"✔ Using REAL yields: {len(data)} rows")
else:
    # simulate to keep end-to-end working (units arbitrary)
    data = seq_df.copy()
    rng = np.random.default_rng(SEED)
    w0,w1,w2 = 2.0,30.0,15.0
    noise = rng.normal(0,3.0, size=len(data))
    data[YIELD_COL] = w0 + w1*data["mean_ndvi"] + w2*data["peak"] + noise
    print("⚠️ No yield CSV provided — simulating yields for model training.")

# Train/validate/test split
if len(data) < 20:
    warnings.warn("Very few samples; metrics may be noisy.")
step_cols = [c for c in data.columns if c.startswith("step_")]
train_df, val_df, test_df = split_df(data, TEST_SIZE, VAL_SIZE, SEED)
X_train, y_train, x_mean, x_std = make_tensors(train_df, step_cols)
X_val  = ((val_df[step_cols].to_numpy(np.float32)-x_mean)/x_std)[:,:,None]
y_val  = val_df[[YIELD_COL]].to_numpy(np.float32)
X_test = ((test_df[step_cols].to_numpy(np.float32)-x_mean)/x_std)[:,:,None]
y_test = test_df[[YIELD_COL]].to_numpy(np.float32)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = train_lstm(X_train, y_train, X_val, y_val, device=device)

# Evaluate
model.eval()
with torch.no_grad():
    yhat = model(torch.from_numpy(X_test).to(device)).cpu().numpy().reshape(-1)
y_true = y_test.reshape(-1)
print(f"\nTest metrics:  R2={r2(y_true,yhat):.3f}  MAE={mae(y_true,yhat):.3f}  RMSE={rmse(y_true,yhat):.3f}")

# Save artifacts
torch.save({"state_dict": model.state_dict(),
            "x_mean": x_mean, "x_std": x_std,
            "config": {"N_STEPS": N_STEPS, "HIDDEN_DIM": HIDDEN_DIM, "NUM_LAYERS": NUM_LAYERS}},
           MODEL_PATH)
pred_df = test_df[["field_id","year"]].copy()
pred_df["y_true"] = y_true; pred_df["y_pred"] = yhat
pred_df.to_csv(OUT_PRED_CSV, index=False)
print(f"✓ Saved model → {MODEL_PATH}")
print(f"✓ Saved predictions → {OUT_PRED_CSV}")

# Quick calibration plot
plt.figure(figsize=(5,5))
plt.scatter(y_true, yhat, alpha=0.7)
lims = [min(y_true.min(), yhat.min()), max(y_true.max(), yhat.max())]
plt.plot(lims, lims)
plt.xlabel("True yield"); plt.ylabel("Predicted yield"); plt.title(f"Pred vs True (R2={r2(y_true,yhat):.2f})")
plt.show()
```
