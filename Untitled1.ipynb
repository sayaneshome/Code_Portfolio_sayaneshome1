```python
# %% [markdown]
# AI-driven crop lifecycle monitoring & yield forecasting from satellite NDVI
# - Extract per-field NDVI time series from mosaics
# - Build seasonal sequences + phenology features
# - Train a PyTorch LSTM to forecast yield
# - Compatible with older scikit-learn (no 'squared' kw) and NumPy 'trapz' deprecation

# %%
import os, re, glob, math, sys, random, warnings
from datetime import datetime
from typing import Optional, Tuple, Dict, Any, List

import numpy as np
import pandas as pd

try:
    from tqdm import tqdm
except Exception:
    tqdm = lambda x, **k: x

# Geospatial stack (optional for DEMO)
try:
    import rasterio
    from rasterio.enums import Resampling
    from rasterio.warp import reproject
    from rasterio.features import rasterize
    import geopandas as gpd
    from affine import Affine
    HAS_GEO = True
except Exception as e:
    HAS_GEO = False
    warnings.warn(f"Geospatial stack not available: {e}")

import matplotlib.pyplot as plt

# Metrics/ML
_SK_OK = True
try:
    from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
    from sklearn.model_selection import train_test_split
except Exception as e:
    _SK_OK = False
    warnings.warn(f"scikit-learn not available; using minimal fallbacks. {e}")

import torch
from torch import nn

# --------------------------
# CONFIG (edit these first)
# --------------------------
MOSAICS_DIR: Optional[str] = None  # e.g., "modis_ndvi_iowa_outputs_offline/mosaics"
PATTERN: str = "NDVI_*.tif"

# Choose **one** mask source:
CDL_PATH: Optional[str]  = None                 # e.g., "/path/to/CDL_IA_2023_30m.tif"
POLY_PATH: Optional[str] = None                 # e.g., "/path/to/fields.shp" (expects 'field_id' if available)

# Optional real yields
YIELD_CSV: Optional[str] = None                 # CSV with columns: field_id, year, yield
YIELD_COL: str = "yield"

# Outputs
OUT_DIR: str = "crop_yield_results"
OUT_TS_CSV: str = "ndvi_timeseries_per_field.csv"
OUT_SEAS_CSV: str = "ndvi_seasonal_sequences.csv"
OUT_PRED_CSV: str = "yield_predictions.csv"
MODEL_PATH: str = "lstm_yield_model.pt"

# NDVI cleaning & options
SCALE: Optional[float] = None
NO_CLIP: bool = False
CLIP_RANGE: Tuple[float, float] = (-1.0, 1.0)
CDL_CLASS: int = 1
ALL_TOUCHED: bool = False

# Season window and resampling
SEASON_DOY_LOW = 100
SEASON_DOY_HIGH = 300
N_STEPS = 32
MIN_POINTS_PER_SEASON = 3

# Training config
TEST_SIZE = 0.2
VAL_SIZE = 0.2
BATCH_SIZE = 32
EPOCHS = 50
LR = 1e-3
HIDDEN_DIM = 64
NUM_LAYERS = 1
PATIENCE = 6

# Random seeds
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)

# --------------- Compat helpers ----------------
def trapz_compat(y, x):
    """Use np.trapezoid if available; else np.trapz."""
    if hasattr(np, "trapezoid"):
        return np.trapezoid(y, x)
    return np.trapz(y, x)

def rmse_compat(y_true, y_pred):
    """RMSE that works with old/new scikit-learn."""
    if _SK_OK:
        try:
            return mean_squared_error(y_true, y_pred, squared=False)
        except TypeError:
            # older scikit-learn
            return float(np.sqrt(mean_squared_error(y_true, y_pred)))
    else:
        # minimal fallback
        y_true = np.asarray(y_true).reshape(-1)
        y_pred = np.asarray(y_pred).reshape(-1)
        return float(np.sqrt(np.mean((y_true - y_pred) ** 2)))

def r2_compat(y_true, y_pred):
    if _SK_OK:
        return float(r2_score(y_true, y_pred))
    else:
        # minimal fallback R^2
        y_true = np.asarray(y_true).reshape(-1)
        y_pred = np.asarray(y_pred).reshape(-1)
        ss_res = np.sum((y_true - y_pred) ** 2)
        ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
        return float(1 - ss_res / ss_tot) if ss_tot > 0 else float("nan")

def mae_compat(y_true, y_pred):
    if _SK_OK:
        return float(mean_absolute_error(y_true, y_pred))
    else:
        y_true = np.asarray(y_true).reshape(-1)
        y_pred = np.asarray(y_pred).reshape(-1)
        return float(np.mean(np.abs(y_true - y_pred)))

# --------------- Utils ----------------
def ensure_dir(p: str):
    os.makedirs(p, exist_ok=True)

def parse_date_from_name(name: str) -> Optional[str]:
    m = re.search(r"NDVI_(\d{8})", name)
    if m:
        return datetime.strptime(m.group(1), "%Y%m%d").date().isoformat()
    return None

def doy_from_iso(iso_date: str) -> int:
    d = datetime.fromisoformat(iso_date)
    return int(d.timetuple().tm_yday)

def get_scale_and_offset(src, user_scale: Optional[float]) -> Tuple[float, float]:
    scale = 1.0; offset = 0.0
    if user_scale is not None:
        return float(user_scale), 0.0
    try:
        if getattr(src, "scales", None) and len(src.scales) > 0 and src.scales[0] not in (None, 1.0):
            scale = float(src.scales[0])
        if getattr(src, "offsets", None) and len(src.offsets) > 0 and src.offsets[0] not in (None, 0.0):
            offset = float(src.offsets[0])
    except Exception:
        pass
    try:
        tags = src.tags()
        if "scale_factor" in tags:
            scale = float(tags["scale_factor"])
        if "add_offset" in tags:
            offset = float(tags["add_offset"])
    except Exception:
        pass
    return scale, offset

def clean_ndvi(arr: np.ndarray, nodata_value, do_clip: bool, low: float, high: float) -> np.ndarray:
    if nodata_value is not None and not (isinstance(nodata_value, float) and math.isnan(nodata_value)):
        arr = np.where(arr == nodata_value, np.nan, arr)
    arr = np.where(arr < -1e19, np.nan, arr)
    if do_clip:
        arr = np.clip(arr, low, high)
    return arr

# --------------- Mask builders ----------------
def build_mask_from_cdl_like(profile: Dict[str, Any], cdl_path: str, corn_code: int) -> np.ndarray:
    with rasterio.open(cdl_path) as src:
        dst = np.zeros((profile["height"], profile["width"]), dtype=np.uint16)
        reproject(
            source=rasterio.band(src, 1),
            destination=dst,
            src_transform=src.transform, src_crs=src.crs,
            dst_transform=profile["transform"], dst_crs=profile["crs"],
            resampling=Resampling.nearest, dst_nodata=0,
        )
    return (dst == corn_code)

def build_field_id_raster(profile: Dict[str, Any], poly_path: str) -> np.ndarray:
    gdf = gpd.read_file(poly_path)
    if gdf.empty:
        raise SystemExit("Polygon file is empty.")
    if gdf.crs is None:
        raise SystemExit("Polygon file has no CRS defined.")

    id_col = None
    for c in ["field_id", "Field_ID", "field", "FID", "id", "ID"]:
        if c in gdf.columns:
            id_col = c; break
    if id_col is None:
        gdf = gdf.reset_index(drop=False).rename(columns={"index": "field_id"})
        id_col = "field_id"

    gg = gdf.to_crs(profile["crs"])
    shapes = [(geom, int(fid)) for geom, fid in zip(gg.geometry, gg[id_col]) if geom is not None and not geom.is_empty]

    fid_arr = rasterize(
        shapes=shapes,
        out_shape=(profile["height"], profile["width"]),
        transform=profile["transform"],
        fill=0, all_touched=ALL_TOUCHED, dtype="int32"
    )
    return fid_arr  # 0 = background

# --------------- Extraction ----------------
def extract_ndvi_timeseries(
    mosaics_dir: str, pattern: str,
    cdl_path: Optional[str], poly_path: Optional[str]
) -> pd.DataFrame:
    tifs = sorted(glob.glob(os.path.join(mosaics_dir, pattern)))
    if not tifs:
        raise SystemExit(f"No mosaics found in {mosaics_dir!r} matching {pattern!r}")

    rows = []
    field_cache = None

    for tif in tqdm(tifs, desc="Extracting per-field NDVI"):
        date_iso = parse_date_from_name(os.path.basename(tif)) or ""
        with rasterio.open(tif) as src:
            arr = src.read(1, masked=False).astype(np.float32)
            scale, offset = get_scale_and_offset(src, SCALE)
            if scale != 1.0 or offset != 0.0:
                arr = arr * scale + offset
            arr = clean_ndvi(arr, src.nodata, not NO_CLIP, CLIP_RANGE[0], CLIP_RANGE[1])

            profile = {"height": src.height, "width": src.width, "crs": src.crs, "transform": src.transform}
            key = (str(src.crs), tuple(src.transform) if hasattr(src.transform, "__iter__") else src.transform, src.width, src.height)

            if field_cache is None or field_cache["key"] != key:
                if poly_path:
                    fid_arr = build_field_id_raster(profile, poly_path)
                    field_cache = {"key": key, "mode": "poly", "fid_arr": fid_arr}
                elif cdl_path:
                    corn_mask = build_mask_from_cdl_like(profile, cdl_path, CDL_CLASS)
                    field_cache = {"key": key, "mode": "cdl", "mask": corn_mask}
                else:
                    field_cache = {"key": key, "mode": "none"}

            if field_cache["mode"] == "poly":
                fid_arr = field_cache["fid_arr"]
                valid = np.isfinite(arr) & (fid_arr > 0)
                if valid.any():
                    s = pd.Series(arr[valid], index=fid_arr[valid])
                    means = s.groupby(level=0).mean()
                    counts = s.groupby(level=0).size()
                    for fid, val in means.items():
                        rows.append({
                            "date": date_iso, "year": int(date_iso[:4]), "doy": doy_from_iso(date_iso),
                            "field_id": int(fid), "ndvi_mean": float(val), "n_pixels": int(counts.loc[fid])
                        })
            elif field_cache["mode"] == "cdl":
                mask = field_cache["mask"]
                valid = np.isfinite(arr) & mask
                ndvi_mean = float(np.nanmean(arr[valid])) if valid.any() else float("nan")
                n_pix = int(valid.sum())
                rows.append({
                    "date": date_iso, "year": int(date_iso[:4]), "doy": doy_from_iso(date_iso),
                    "field_id": 1, "ndvi_mean": ndvi_mean, "n_pixels": n_pix
                })
            else:
                valid = np.isfinite(arr)
                ndvi_mean = float(np.nanmean(arr[valid])) if valid.any() else float("nan")
                n_pix = int(valid.sum())
                rows.append({
                    "date": date_iso, "year": int(date_iso[:4]), "doy": doy_from_iso(date_iso),
                    "field_id": 1, "ndvi_mean": ndvi_mean, "n_pixels": n_pix
                })

    df = pd.DataFrame(rows).sort_values(["field_id", "date"])
    return df

# --------------- Season building ----------------
def build_season_sequences(df_long: pd.DataFrame, n_steps=N_STEPS) -> pd.DataFrame:
    seq_rows = []
    for (fid, yr), g in df_long.groupby(["field_id", "year"]):
        gg = g[(g["doy"] >= SEASON_DOY_LOW) & (g["doy"] <= SEASON_DOY_HIGH)].sort_values("doy")
        gg = gg.dropna(subset=["ndvi_mean"])
        if len(gg) < MIN_POINTS_PER_SEASON:
            continue

        xs = gg["doy"].to_numpy()
        ys = gg["ndvi_mean"].to_numpy()
        target = np.linspace(SEASON_DOY_LOW, SEASON_DOY_HIGH, n_steps)

        uniq_xs, idx = np.unique(xs, return_index=True)
        uniq_ys = ys[idx]
        if len(uniq_xs) == 1:
            seq = np.repeat(uniq_ys[0], n_steps)
        else:
            seq = np.interp(target, uniq_xs, uniq_ys)

        auc = float(trapz_compat(seq, target))
        peak = float(np.nanmax(seq))
        meanv = float(np.nanmean(seq))

        row = {"field_id": int(fid), "year": int(yr), "AUC": auc, "peak": peak, "mean_ndvi": meanv}
        for i, v in enumerate(seq):
            row[f"step_{i:02d}"] = float(v)
        seq_rows.append(row)

    out = pd.DataFrame(seq_rows).sort_values(["field_id", "year"])
    return out

# --------------- Yield handling ----------------
def attach_or_simulate_yield(df_seq: pd.DataFrame, yield_csv: Optional[str]) -> Tuple[pd.DataFrame, bool]:
    if yield_csv and os.path.isfile(yield_csv):
        ydf = pd.read_csv(yield_csv)
        need = {"field_id", "year", YIELD_COL}
        if not need.issubset(set(ydf.columns)):
            raise SystemExit(f"Yield CSV must have columns {need}, found {set(ydf.columns)}")
        out = df_seq.merge(ydf[["field_id", "year", YIELD_COL]], on=["field_id", "year"], how="inner")
        if out.empty:
            raise SystemExit("No (field_id, year) overlap between sequences and yield CSV.")
        return out, False

    # simulate yields for demo/training
    df = df_seq.copy()
    rng = np.random.default_rng(SEED)
    w0, w1, w2 = 2.0, 30.0, 15.0
    noise = rng.normal(0, 3.0, size=len(df))
    df[YIELD_COL] = w0 + w1 * df["mean_ndvi"] + w2 * df["peak"] + noise
    return df, True

# --------------- LSTM Model ----------------
class YieldLSTM(nn.Module):
    def __init__(self, input_dim=1, hidden_dim=HIDDEN_DIM, num_layers=NUM_LAYERS):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers=num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1)
    def forward(self, x):
        out, _ = self.lstm(x)
        last = out[:, -1, :]
        y = self.fc(last)
        return y

def make_tensors(df: pd.DataFrame, step_cols: List[str]):
    X = df[step_cols].to_numpy(dtype=np.float32)
    x_mean = X.mean(axis=0, keepdims=True)
    x_std = X.std(axis=0, keepdims=True) + 1e-8
    Xn = (X - x_mean) / x_std
    Xn = Xn[:, :, None]
    y = df[[YIELD_COL]].to_numpy(dtype=np.float32)
    return Xn, y, x_mean.astype(np.float32), x_std.astype(np.float32)

def _train_val_split(data, test_size, val_size, seed):
    if _SK_OK:
        from sklearn.model_selection import train_test_split as sk_split
        train_df, test_df = sk_split(data, test_size=test_size, random_state=seed)
        train_df, val_df  = sk_split(train_df, test_size=val_size, random_state=seed)
        return train_df, val_df, test_df
    # minimal fallback
    idx = np.arange(len(data))
    rng = np.random.default_rng(seed); rng.shuffle(idx)
    n_test = int(len(data)*test_size); n_val = int((len(data)-n_test)*val_size)
    test_idx = idx[:n_test]; val_idx = idx[n_test:n_test+n_val]; train_idx = idx[n_test+n_val:]
    return data.iloc[train_idx], data.iloc[val_idx], data.iloc[test_idx]

def train_lstm(X_train, y_train, X_val, y_val, device="cpu"):
    model = YieldLSTM().to(device)
    opt = torch.optim.Adam(model.parameters(), lr=LR)
    loss_fn = nn.MSELoss()

    def batches(X, y, bs):
        n = len(X)
        idx = np.arange(n)
        for i in range(0, n, bs):
            j = idx[i:i+bs]
            yield X[j], y[j]

    best_val = float("inf")
    best_state = None
    patience = PATIENCE
    for epoch in range(1, EPOCHS+1):
        model.train()
        train_losses = []
        for xb, yb in batches(X_train, y_train, BATCH_SIZE):
            xb = torch.from_numpy(xb).to(device)
            yb = torch.from_numpy(yb).to(device)
            opt.zero_grad()
            pred = model(xb)
            loss = loss_fn(pred, yb)
            loss.backward()
            opt.step()
            train_losses.append(loss.item())

        model.eval()
        with torch.no_grad():
            xv = torch.from_numpy(X_val).to(device)
            yv = torch.from_numpy(y_val).to(device)
            pv = model(xv)
            vloss = loss_fn(pv, yv).item()

        print(f"Epoch {epoch:03d}  train_loss={np.mean(train_losses):.4f}  val_loss={vloss:.4f}  (patience {patience})")

        if vloss < best_val - 1e-6:
            best_val = vloss
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            patience = PATIENCE
        else:
            patience -= 1
            if patience == 0:
                print("Early stopping.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)
    return model

# --------------- Orchestration ----------------
def run_full_pipeline():
    ensure_dir(OUT_DIR)
    real_ok = HAS_GEO and MOSAICS_DIR and os.path.isdir(MOSAICS_DIR) and \
              len(glob.glob(os.path.join(MOSAICS_DIR, PATTERN))) > 0 and (CDL_PATH or POLY_PATH)

    if not real_ok:
        print("⚠️  Running DEMO mode (synthetic data). Set MOSAICS_DIR + (CDL_PATH or POLY_PATH) to use real data.\n")
        # DEMO: 5 fields × 3 years of synthetic seasonal NDVI
        fields = [101,102,103,104,105]
        years = [2021, 2022, 2023]
        target = np.linspace(SEASON_DOY_LOW, SEASON_DOY_HIGH, N_STEPS)
        seq_rows = []
        rng = np.random.default_rng(SEED)
        for f in fields:
            base_peak = rng.uniform(0.6, 0.85)
            for y in years:
                center = rng.uniform(0.55, 0.65) * (SEASON_DOY_HIGH-SEASON_DOY_LOW) + SEASON_DOY_LOW
                width = rng.uniform(40, 70)
                seq = base_peak * np.exp(-0.5*((target-center)/width)**2)
                seq += rng.normal(0, 0.02, size=len(seq))
                seq = np.clip(seq, 0, 1)
                auc = float(trapz_compat(seq, target))
                peak = float(np.max(seq))
                meanv = float(np.mean(seq))
                row = {"field_id": f, "year": y, "AUC": auc, "peak": peak, "mean_ndvi": meanv}
                for i, v in enumerate(seq):
                    row[f"step_{i:02d}"] = float(v)
                seq_rows.append(row)
        seq_df = pd.DataFrame(seq_rows).sort_values(["field_id","year"])
        seq_df, sim = attach_or_simulate_yield(seq_df, None)
        df_long = None
    else:
        df_long = extract_ndvi_timeseries(MOSAICS_DIR, PATTERN, CDL_PATH, POLY_PATH)
        df_long.to_csv(os.path.join(OUT_DIR, OUT_TS_CSV), index=False)
        print(f"✓ Wrote per-date NDVI timeseries → {os.path.join(OUT_DIR, OUT_TS_CSV)}  ({len(df_long)} rows)")
        seq_df = build_season_sequences(df_long, N_STEPS)
        if seq_df.empty:
            raise SystemExit("No valid seasonal sequences (check season DOY window or data coverage).")
        seq_df.to_csv(os.path.join(OUT_DIR, OUT_SEAS_CSV), index=False)
        print(f"✓ Wrote seasonal sequences → {os.path.join(OUT_DIR, OUT_SEAS_CSV)}  ({len(seq_df)} rows)")
        seq_df, sim = attach_or_simulate_yield(seq_df, YIELD_CSV)
        print("✓ Using", "SIMULATED" if sim else "REAL", "yields")

    # Prepare data
    step_cols = [c for c in seq_df.columns if c.startswith("step_")]
    data = seq_df.dropna(subset=step_cols + [YIELD_COL]).reset_index(drop=True)
    if len(data) < 20:
        warnings.warn("Very few samples; metrics will be noisy.")

    # Splits
    train_df, val_df, test_df = _train_val_split(data, TEST_SIZE, VAL_SIZE, SEED)

    X_train, y_train, x_mean, x_std = make_tensors(train_df, step_cols)
    X_val  = ((val_df[step_cols].to_numpy(dtype=np.float32) - x_mean) / x_std)[:, :, None]
    y_val  = val_df[[YIELD_COL]].to_numpy(dtype=np.float32)
    X_test = ((test_df[step_cols].to_numpy(dtype=np.float32) - x_mean) / x_std)[:, :, None]
    y_test = test_df[[YIELD_COL]].to_numpy(dtype=np.float32)

    device = "cuda" if torch.cuda.is_available() else "cpu"
    model = train_lstm(X_train, y_train, X_val, y_val, device=device)

    # Evaluate
    model.eval()
    with torch.no_grad():
        yhat = model(torch.from_numpy(X_test).to(device)).cpu().numpy().reshape(-1)
    y_true = y_test.reshape(-1)
    r2  = r2_compat(y_true, yhat)
    mae = mae_compat(y_true, yhat)
    rmse = rmse_compat(y_true, yhat)
    print(f"\nTest metrics:  R2={r2:.3f}  MAE={mae:.3f}  RMSE={rmse:.3f}")

    # Save model + predictions
    ensure_dir(OUT_DIR)
    torch.save({"state_dict": model.state_dict(), "x_mean": x_mean, "x_std": x_std,
                "config": {"N_STEPS": N_STEPS, "HIDDEN_DIM": HIDDEN_DIM, "NUM_LAYERS": NUM_LAYERS}}, 
               os.path.join(OUT_DIR, MODEL_PATH))

    pred_df = test_df[["field_id","year"]].copy()
    pred_df["y_true"] = y_true
    pred_df["y_pred"] = yhat
    pred_df.to_csv(os.path.join(OUT_DIR, OUT_PRED_CSV), index=False)
    print(f"✓ Saved model → {os.path.join(OUT_DIR, MODEL_PATH)}")
    print(f"✓ Saved predictions → {os.path.join(OUT_DIR, OUT_PRED_CSV)}")

    # Plots
    plt.figure(figsize=(5,5))
    plt.scatter(y_true, yhat, alpha=0.7)
    lims = [min(y_true.min(), yhat.min()), max(y_true.max(), yhat.max())]
    plt.plot(lims, lims)
    plt.xlabel("True yield"); plt.ylabel("Predicted yield"); plt.title(f"Pred vs True (R2={r2:.2f})")
    plt.show()

    if df_long is not None and "field_id" in df_long.columns:
        sample_fid = int(df_long["field_id"].mode().iloc[0])
        dd = df_long[df_long["field_id"]==sample_fid]
        plt.figure(figsize=(8,3))
        for yr, g in dd.groupby("year"):
            m = (g["doy"]>=SEASON_DOY_LOW)&(g["doy"]<=SEASON_DOY_HIGH)
            plt.plot(g.loc[m,"doy"], g.loc[m,"ndvi_mean"], marker="o", linestyle="-", label=str(yr))
        plt.axvspan(SEASON_DOY_LOW, SEASON_DOY_HIGH, color="0.9")
        plt.title(f"Field {sample_fid} NDVI (season window shaded)")
        plt.xlabel("DOY"); plt.ylabel("NDVI")
        plt.legend(ncol=3, fontsize=8)
        plt.show()

# ---------------- Run ----------------
if __name__ == "__main__" or True:
    run_full_pipeline()
```

```python
# %% Real-data bootstrap (auto-detect paths, set globals, and run the pipeline)
import os, glob, re, numpy as np
from pathlib import Path

# ---------- SEARCH ROOTS (add your project folders here if needed) ----------
SEARCH_ROOTS = [Path("."), Path(".."), Path("data"), Path("outputs"), Path("mosaics")]

def _find_mosaic_dir():
    candidates = []
    for root in SEARCH_ROOTS:
        for p in root.rglob("NDVI_*.tif"):
            candidates.append(p.parent.resolve())
    uniq = []
    for d in candidates:
        if d not in uniq:
            uniq.append(d)
    return uniq[0] if uniq else None

def _peek_raster_info(sample_tif):
    import rasterio
    with rasterio.open(sample_tif) as src:
        arr = src.read(1, masked=False)
        nodata = src.nodata
        dtype = arr.dtype
        # fast min/max on a small subset
        flat = arr.reshape(-1)
        sample = flat[::max(1, flat.size // 500000)]
        # ignore huge negative fill
        sample = sample[sample > -1e19]
        vmin = float(np.nanmin(sample)) if sample.size else np.nan
        vmax = float(np.nanmax(sample)) if sample.size else np.nan
        return dtype, vmin, vmax, nodata

def _auto_scale(dtype, vmin, vmax):
    """
    Heuristic:
      - float NDVI mostly within [-1,1] -> scale=1.0
      - int NDVI often in [0..10000] or [-10000..10000] -> scale=0.0001
    """
    if np.issubdtype(dtype, np.floating):
        return 1.0
    # int
    if vmax is not np.nan and (vmax > 1.5 and vmax <= 20000):
        return 1e-4
    # conservative default
    return 1.0

def _find_mask():
    # Prefer polygons if we see obvious field/corn layers
    poly_hits = []
    for root in SEARCH_ROOTS:
        poly_hits += list(root.rglob("*.shp"))
        poly_hits += list(root.rglob("*.geojson"))
    # prefer names containing 'field' or 'corn'
    good = [p for p in poly_hits if re.search(r"(field|corn|parcel|farm)", p.name, re.I)]
    if good:
        return ("poly", good[0].resolve())

    # Else try CDL
    cdl_hits = []
    for root in SEARCH_ROOTS:
        cdl_hits += list(root.rglob("*CDL*30m*.tif"))
        cdl_hits += list(root.rglob("*CDL*.tif"))
        cdl_hits += list(root.rglob("CDL_*.tif"))
    if cdl_hits:
        return ("cdl", cdl_hits[0].resolve())

    return (None, None)

def _find_yield_csv():
    csvs = []
    for root in SEARCH_ROOTS:
        csvs += list(root.rglob("*.csv"))
    # prefer files with 'yield' in the name
    good = [p for p in csvs if re.search(r"yield", p.name, re.I)]
    return good[0].resolve() if good else None

# --------- 1) Locate mosaics ---------
mosaic_dir = _find_mosaic_dir()
if mosaic_dir is None:
    raise SystemExit(
        "Couldn’t find any NDVI_*.tif mosaics. "
        "Set MOSAICS_DIR manually (folder containing NDVI_YYYYMMDD.tif files)."
    )

tifs = sorted(glob.glob(str(mosaic_dir / "NDVI_*.tif")))
if not tifs:
    raise SystemExit(f"No NDVI_*.tif found in {mosaic_dir}")

dtype, vmin, vmax, nodata = _peek_raster_info(tifs[0])
auto_scale = _auto_scale(dtype, vmin, vmax)

# --------- 2) Locate mask (polygons preferred) ---------
mask_kind, mask_path = _find_mask()
if mask_kind is None:
    print("⚠️  No polygons/CDL found by auto-search.")
    print("    Provide *one* of the following in your notebook globals, then re-run this cell:")
    print("      POLY_PATH = '/absolute/path/to/fields.shp'   # preferred (expects 'field_id' column if present)")
    print("      CDL_PATH  = '/absolute/path/to/CDL_IA_2023_30m.tif'  # and set CDL_CLASS=1 for corn")
    raise SystemExit

# --------- 3) Try to locate a yield CSV (optional) ---------
yield_csv = _find_yield_csv()
yield_col = "yield"
if yield_csv is not None:
    import pandas as pd
    ydf = pd.read_csv(yield_csv, nrows=1)
    cols = [c for c in ydf.columns]
    # pick a column that looks like yield
    cand = [c for c in cols if re.search(r"yield|yld", c, re.I)]
    if cand:
        yield_col = cand[0]
    # else leave as default; pipeline will simulate if missing

# --------- 4) Set globals for the pipeline cell you already ran ---------
globals()["MOSAICS_DIR"] = str(mosaic_dir)
globals()["PATTERN"] = "NDVI_*.tif"

if mask_kind == "poly":
    globals()["POLY_PATH"] = str(mask_path)
    globals()["CDL_PATH"] = None
else:
    globals()["CDL_PATH"] = str(mask_path)
    globals()["POLY_PATH"] = None
    globals()["CDL_CLASS"] = 1  # 🟡 adjust if your CDL uses a different corn class

# Scale (override only if pipeline couldn’t read scale from metadata)
globals()["SCALE"] = auto_scale  # set 1e-4 automatically for int NDVI
globals()["NO_CLIP"] = False
globals()["CLIP_RANGE"] = (-1.0, 1.0)

# Optional yields
globals()["YIELD_CSV"] = str(yield_csv) if yield_csv else None
globals()["YIELD_COL"] = yield_col  # the pipeline expects this name
# If your CSV column isn’t literally 'yield', the pipeline uses YIELD_COL

# Quality-of-life prints
print("✔ Mosaics dir :", globals()["MOSAICS_DIR"])
print("✔ Sample tif  :", tifs[0])
print(f"   dtype={dtype}, vmin≈{vmin:.4f}, vmax≈{vmax:.4f}, nodata={nodata}, auto SCALE={globals()['SCALE']}")
print(f"✔ Mask        : {mask_kind.upper()} → {mask_path}")
print(f"✔ Yields      : {globals()['YIELD_CSV']} (column='{globals()['YIELD_COL']}')")

# --------- 5) Run your full pipeline (already defined in the previous cell) ---------
try:
    run_full_pipeline()
except NameError:
    raise SystemExit("run_full_pipeline() is not defined in this kernel. "
                     "Please run the big pipeline cell first, then re-run this cell.")
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
AI-driven crop lifecycle monitoring & yield forecasting (county-level, USA)
- Downloads MODIS NDVI via Google Earth Engine (masked to crop using USDA CDL)
- Builds per-year NDVI seasonal sequences (growing season DOY 80-320)
- Summarizes features (AUC, peak, mean_ndvi)
- Fetches USDA NASS county yields for the same crop/years
- Trains a tiny LSTM on sequences to predict yield
- Saves: crop_yield_results/ndvi_timeseries_per_field.csv
         crop_yield_results/ndvi_seasonal_sequences.csv
         crop_yield_results/yield_predictions.csv
"""
import os, sys, json, time, argparse, math, itertools, warnings, tempfile
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
from tqdm import tqdm
import requests

# ML
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"
import tensorflow as tf
from tensorflow.keras import layers, models

# Plot (optional; no figure saving by default)
import matplotlib.pyplot as plt

# Earth Engine
try:
    import ee
    import geemap
except Exception as e:
    print("ERROR: earthengine-api/geemap not installed. Install deps first:\n"
          "  pip install earthengine-api geemap pandas numpy requests tqdm tensorflow scikit-learn matplotlib")
    sys.exit(1)

OUT_DIR  = "crop_yield_results"
TS_CSV   = os.path.join(OUT_DIR, "ndvi_timeseries_per_field.csv")
SEAS_CSV = os.path.join(OUT_DIR, "ndvi_seasonal_sequences.csv")
PRED_CSV = os.path.join(OUT_DIR, "yield_predictions.csv")

# -------------------------- Utils --------------------------

def ensure_dir(d):
    os.makedirs(d, exist_ok=True)

def year_doys(start_doy=80, end_doy=320, step=16):
    """MOD13Q1 is 16-day cadence. Return approximate DOY midpoints."""
    return list(range(start_doy, end_doy+1, step))

def trapezoid_auc(y_vals, x_step=16):
    """Simple trapezoidal AUC over equispaced x (days)."""
    y = np.asarray(y_vals, float)
    if len(y) < 2:
        return float(np.nan)
    return float(np.trapz(y, dx=x_step))

def to_fips(state_abbr, county_name):
    """
    Resolve county FIPS using US Census API-free CSV snapshot via geemap assets.
    Fallback: TIGER via geemap.
    """
    try:
        # geemap has a counties feature collection
        counties = ee.FeatureCollection("TIGER/2018/Counties").filter(ee.Filter.eq('STATEFP', state_to_fips(state_abbr)))
        county = counties.filter(ee.Filter.eq('NAME', county_name)).first()
        fips = county.get('GEOID').getInfo()
        return fips
    except Exception:
        return None

def state_to_fips(state_abbr):
    table = {
        'AL':'01','AK':'02','AZ':'04','AR':'05','CA':'06','CO':'08','CT':'09','DE':'10','DC':'11','FL':'12',
        'GA':'13','HI':'15','ID':'16','IL':'17','IN':'18','IA':'19','KS':'20','KY':'21','LA':'22','ME':'23',
        'MD':'24','MA':'25','MI':'26','MN':'27','MS':'28','MO':'29','MT':'30','NE':'31','NV':'32','NH':'33',
        'NJ':'34','NM':'35','NY':'36','NC':'37','ND':'38','OH':'39','OK':'40','OR':'41','PA':'42','RI':'44',
        'SC':'45','SD':'46','TN':'47','TX':'48','UT':'49','VT':'50','VA':'51','WA':'53','WV':'54','WI':'55','WY':'56'
    }
    return table[state_abbr.upper()]

def get_county_geom(state_abbr, county_name):
    """Return county ee.Geometry and FIPS."""
    statefp = state_to_fips(state_abbr)
    fc = ee.FeatureCollection("TIGER/2018/Counties") \
            .filter(ee.Filter.eq("STATEFP", statefp)) \
            .filter(ee.Filter.eq("NAME", county_name))
    f = fc.first()
    if f is None:
        raise RuntimeError(f"County not found: {county_name}, {state_abbr}")
    geoid = f.get("GEOID").getInfo()
    geom = f.geometry()
    return geom, geoid

def init_ee():
    try:
        ee.Initialize(project=None)
    except Exception:
        print("🔐 First-time Earth Engine auth needed. A browser window may open.")
        ee.Authenticate()
        ee.Initialize(project=None)

# ---------------------- Data acquisition ----------------------

def cdl_crop_mask(year, crop_name="CORN"):
    """
    USDA/NASS/CDL crop codes:
      Corn=1, Soybeans=5, Winter Wheat=24, Spring Wheat=23, Alfalfa=36, Cotton=2, etc.
    """
    crop_map = {
        "CORN":1, "SOYBEANS":5, "WINTER WHEAT":24, "SPRING WHEAT":23,
        "COTTON":2, "ALFALFA":36, "RICE":3, "BARLEY":21, "SORGHUM":4
    }
    code = crop_map.get(crop_name.upper(), 1)
    cdl = ee.ImageCollection("USDA/NASS/CDL") \
            .filter(ee.Filter.eq('year', year)).first().select('cropland')
    mask = cdl.eq(code)
    return mask

def modis_ndvi_collection(year):
    """
    MOD13Q1.006 NDVI 16-day (250m). Alternatively MOD13A2 (1km).
    Scale: 250m.
    """
    start = ee.Date.fromYMD(year, 1, 1).advance(79, 'day')   # ~DOY 80
    end   = ee.Date.fromYMD(year, 1, 1).advance(320, 'day')  # ~DOY 320
    col = ee.ImageCollection("MODIS/006/MOD13Q1") \
            .filterDate(start, end) \
            .select('NDVI')
    return col

def county_mean_ndvi_series(geom, year, crop_name):
    """
    Return list of dicts: [{'year':y, 'doy':doy, 'ndvi':val}, ...]
    """
    mask = cdl_crop_mask(year, crop_name)
    col = modis_ndvi_collection(year).map(lambda img: img.updateMask(mask))
    # Sort by date
    col = col.sort('system:time_start')
    # Reduce each image over the county polygon
    imgs = col.toList(col.size())
    n = imgs.size().getInfo()
    out = []
    for i in range(n):
        img = ee.Image(imgs.get(i))
        dt  = ee.Date(img.get('system:time_start'))
        doy = dt.getRelative('day', 'year').add(1).getInfo()
        mean = img.reduceRegion(
            reducer=ee.Reducer.mean(),
            geometry=geom,
            scale=250,
            maxPixels=1e13
        ).get('NDVI')
        val = mean.getInfo()
        # MODIS NDVI is scaled by 1e4
        if val is None:
            ndvi = np.nan
        else:
            ndvi = float(val) / 1e4
        out.append({"year": int(year), "doy": int(doy), "ndvi": ndvi})
    return out

def fetch_ndvi_timeseries(state, county, crop, years):
    geom, geoid = get_county_geom(state, county)
    records = []
    for y in years:
        seq = county_mean_ndvi_series(geom, y, crop)
        for r in seq:
            r.update({"field_id": geoid})  # use county FIPS as field_id
            records.append(r)
    df = pd.DataFrame(records)
    # sort and drop all-nan years if any
    df.sort_values(["field_id","year","doy"], inplace=True)
    # Keep only regular 16-d cadence DOYs within [80,320]
    df = df[(df["doy"]>=80) & (df["doy"]<=320)]
    return df, geoid

# ---------------------- NASS yields ----------------------

def nass_county_yield(state_abbr, county_name, crop_name, years, api_key=None):
    """
    Pull county-level yield (BU / ACRE) for given crop.
    Commodity names must match NASS: e.g., 'CORN', 'SOYBEANS', 'WHEAT'.
    """
    if api_key is None:
        api_key = os.environ.get("NASS_API_KEY", None)
    if api_key is None:
        raise RuntimeError("NASS_API_KEY not set. Get a free key at https://quickstats.nass.usda.gov/api")

    commodity = crop_name.upper()
    params_base = {
        "key": api_key,
        "source_desc": "SURVEY",
        "sector_desc": "CROPS",
        "group_desc": "FIELD CROPS",
        "commodity_desc": commodity,
        "statisticcat_desc": "YIELD",
        "unit_desc": "BU / ACRE",
        "agg_level_desc": "COUNTY",
        "state_alpha": state_abbr.upper(),
        "county_name": county_name.upper(),  # NASS expects uppercase county name without " County"
        "format": "JSON",
    }

    dfs = []
    for y in years:
        params = dict(params_base)
        params["year"] = str(y)
        url = "https://quickstats.nass.usda.gov/api/api_GET/"
        r = requests.get(url, params=params, timeout=60)
        r.raise_for_status()
        data = r.json()
        rows = data.get("data", [])
        if len(rows)==0:
            # Some counties may use "YIELD, MEASURED IN BU / ACRE" variation. Try unit_desc fallback.
            params2 = dict(params_base); params2["year"]=str(y); params2["unit_desc"]=""
            r2 = requests.get(url, params=params2, timeout=60); r2.raise_for_status()
            rows = r2.json().get("data", [])
        if len(rows)>0:
            # There may be multiple practice types; pick ALL PRODUCTION PRACTICES if present, else first
            # Prefer "PRACTICE_DESC": "ALL PRODUCTION PRACTICES"
            chosen = None
            for rr in rows:
                if rr.get("practice_desc","").upper()=="ALL PRODUCTION PRACTICES":
                    chosen = rr; break
            if chosen is None:
                chosen = rows[0]
            val = chosen.get("Value", "").replace(",","").strip()
            try:
                yld = float(val)
            except:
                yld = np.nan
            dfs.append({"year": int(y), "y_true": yld})
        else:
            dfs.append({"year": int(y), "y_true": np.nan})
    return pd.DataFrame(dfs)

# ---------------------- Features & Model ----------------------

def build_seasonal_features(ts_df):
    """
    Input: rows (field_id, year, doy, ndvi)
    Output: seasonal feature table with sequence (as list) and summary stats.
    """
    feats = []
    for (fid, yr), grp in ts_df.groupby(["field_id","year"]):
        g = grp.sort_values("doy")
        seq = g["ndvi"].to_numpy()
        # Handle missing by linear interpolation then fill
        s = pd.Series(seq)
        s = s.interpolate(limit_direction="both")
        seq = s.fillna(s.mean()).to_numpy()
        AUC = trapezoid_auc(seq, x_step=16)
        peak = float(np.nanmax(seq)) if len(seq)>0 else np.nan
        mean_ndvi = float(np.nanmean(seq)) if len(seq)>0 else np.nan
        feats.append({
            "field_id": fid, "year": int(yr),
            "AUC": AUC, "peak": peak, "mean_ndvi": mean_ndvi,
            "sequence": json.dumps([float(x) for x in seq])
        })
    df = pd.DataFrame(feats)
    return df

def make_lstm_dataset(seas_df, window_len=None):
    """
    Convert JSON sequences to [n, T, 1] array, pad/truncate to common length.
    """
    seqs = [np.array(json.loads(s), dtype=float) for s in seas_df["sequence"]]
    T = max(len(s) for s in seqs) if window_len is None else window_len
    X = np.zeros((len(seqs), T, 1), dtype=np.float32)
    for i, s in enumerate(seqs):
        if len(s)>=T:
            X[i,:,0] = s[:T]
        else:
            X[i,:len(s),0] = s
            X[i,len(s):,0] = s[-1] if len(s)>0 else 0.0
    # Simple standardization per time step
    mu = X.mean(axis=0, keepdims=True)
    sd = X.std(axis=0, keepdims=True) + 1e-6
    Xz = (X - mu)/sd
    return Xz

def build_lstm(input_timesteps):
    model = models.Sequential([
        layers.Input(shape=(input_timesteps,1)),
        layers.Masking(mask_value=0.0),
        layers.LSTM(64, return_sequences=False),
        layers.Dense(32, activation="relu"),
        layers.Dense(1)  # yield prediction (BU/ACRE)
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(1e-3),
                  loss="mse",
                  metrics=[tf.keras.metrics.MeanAbsoluteError(name="mae")])
    return model

def r2_score(y, yhat):
    y = np.asarray(y).reshape(-1)
    yhat = np.asarray(yhat).reshape(-1)
    ss_res = np.sum((y - yhat)**2)
    ss_tot = np.sum((y - y.mean())**2)
    return float(1 - ss_res/ss_tot) if ss_tot > 0 else np.nan

# ---------------------- Main ----------------------

def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("--state", required=True, help="State abbreviation, e.g., IA")
    ap.add_argument("--county", required=True, help='County name, e.g., "Story"')
    ap.add_argument("--crop", default="CORN", help="Crop (NASS/CDL names): CORN, SOYBEANS, WINTER WHEAT, etc.")
    ap.add_argument("--start_year", type=int, default=2016)
    ap.add_argument("--end_year", type=int, default=2023)
    ap.add_argument("--epochs", type=int, default=200)
    ap.add_argument("--seed", type=int, default=42)
    args = ap.parse_args()

    np.random.seed(args.seed)
    tf.random.set_seed(args.seed)

    ensure_dir(OUT_DIR)

    # ----- Initialize Earth Engine
    init_ee()

    years = list(range(args.start_year, args.end_year+1))
    print(f"▶ Downloading NDVI for {args.county} County, {args.state}, crop={args.crop}, years={years}")

    # NDVI time series
    ts_df, geoid = fetch_ndvi_timeseries(args.state, args.county, args.crop, years)
    if ts_df.empty:
        raise RuntimeError("No NDVI samples returned (check county/crop/year ranges).")
    # Save timeseries
    # Conform columns to your checker: field_id, year, doy, ndvi
    ts_df = ts_df[["field_id","year","doy","ndvi"]].copy()
    ts_df.to_csv(TS_CSV, index=False)
    print(f"✓ Saved timeseries: {TS_CSV} ({len(ts_df)} rows)")

    # Seasonal sequences + features
    seas_df = build_seasonal_features(ts_df)
    # Save a flat version (replace 'sequence' JSON with comma-joined string for readability)
    seas_save = seas_df.copy()
    seas_save["sequence"] = seas_save["sequence"].apply(lambda s: s)  # keep JSON as-is
    seas_save[["field_id","year","AUC","peak","mean_ndvi","sequence"]].to_csv(SEAS_CSV, index=False)
    print(f"✓ Saved seasonal features: {SEAS_CSV} ({len(seas_df)} years)")

    # Ground truth yields
    print("▶ Fetching USDA NASS county yields …")
    yld_df = nass_county_yield(args.state, args.county, args.crop, years)
    yld_df["field_id"] = geoid
    df = pd.merge(seas_df, yld_df, on=["field_id","year"], how="inner")
    n_before = len(df)
    df = df.dropna(subset=["y_true"])
    if len(df) < 5:
        print("⚠️ Very few yield labels returned by NASS for the requested window. "
              "Consider widening the year range.")
    print(f"✓ Yield labels: {len(df)}/{n_before} years usable.")

    # Split train/test: last year as test (if available); else 80/20 split
    df = df.sort_values("year")
    if len(df)>=2:
        test_year = int(df["year"].max())
        tr = df[df["year"] < test_year].copy()
        te = df[df["year"] == test_year].copy()
    else:
        # fallback
        m = int(0.8*len(df))
        tr = df.iloc[:m].copy()
        te = df.iloc[m:].copy()

    # Build model inputs
    Xtr = make_lstm_dataset(tr, window_len=None)
    Xte = make_lstm_dataset(te, window_len=Xtr.shape[1])
    ytr = tr["y_true"].to_numpy().astype(np.float32)
    yte = te["y_true"].to_numpy().astype(np.float32)

    model = build_lstm(input_timesteps=Xtr.shape[1])
    early = tf.keras.callbacks.EarlyStopping(monitor="val_mae", patience=20, restore_best_weights=True)
    hist = model.fit(Xtr, ytr, validation_split=0.2, epochs=args.epochs, verbose=0, callbacks=[early], batch_size=8)

    # Predict
    yhat_tr = model.predict(Xtr, verbose=0).reshape(-1)
    yhat_te = model.predict(Xte, verbose=0).reshape(-1)

    # Metrics
    def _mae(a,b): return float(np.mean(np.abs(a-b)))
    def _rmse(a,b): return float(np.sqrt(np.mean((a-b)**2)))
    R2_tr, MAE_tr, RMSE_tr = r2_score(ytr,yhat_tr), _mae(ytr,yhat_tr), _rmse(ytr,yhat_tr)
    R2_te, MAE_te, RMSE_te = r2_score(yte,yhat_te), _mae(yte,yhat_te), _rmse(yte,yhat_te)

    print(f"Training  → R2={R2_tr:.3f}  MAE={MAE_tr:.3f}  RMSE={RMSE_tr:.3f}")
    print(f"Test ({int(te['year'].unique()[0])}) → R2={R2_te:.3f}  MAE={MAE_te:.3f}  RMSE={RMSE_te:.3f}")

    # Save predictions in the format your checker expects
    # Multiple rows allowed if you later add more counties ("fields").
    pred_rows = []
    for (fid, yr), grp in te.groupby(["field_id","year"]):
        yhat = float(yhat_te[list(te.index.get_indexer(grp.index))[0]])
        ytru = float(grp["y_true"].iloc[0])
        pred_rows.append({"field_id": fid, "year": int(yr), "y_pred": yhat, "y_true": ytru})
    pred_df = pd.DataFrame(pred_rows)
    pred_df["resid"] = pred_df["y_pred"] - pred_df["y_true"]
    pred_df.to_csv(PRED_CSV, index=False)
    print(f"✓ Saved predictions: {PRED_CSV}")

    # Optional quick plot
    try:
        plt.figure(figsize=(5,5))
        plt.scatter(yte, yhat_te, alpha=0.8)
        lo, hi = float(min(yte.min(), yhat_te.min())), float(max(yte.max(), yhat_te.max()))
        plt.plot([lo,hi],[lo,hi])
        plt.xlabel("True yield (bu/acre)"); plt.ylabel("Predicted yield")
        plt.title(f"{args.county} County {args.state}: Test R2={R2_te:.2f}")
        plt.tight_layout()
        plt.show()
    except Exception as e:
        print("(Plot skipped:", e, ")")

    print("\nAll done. Now run your diagnostic script against:", OUT_DIR)

if __name__ == "__main__":
    main()
```
